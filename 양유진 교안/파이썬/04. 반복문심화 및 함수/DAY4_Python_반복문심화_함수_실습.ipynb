### 2. 함수 정답

````python
# 실습 2-1
def average(values):
    return sum(values) / len(values)

sales = [120000, 150000, 90000, 285000]
avg = average(sales)
print(f"평균 매출: {avg:,.0f}원")
````

````python
# 실습 2-2
def show_total(a, b):
    print(a + b)              # 화면에 보여 주기만 한다

def plus(a, b):
    return a + b              # 값을 돌려준다

result = show_total(2, 3)
print(result)                 # return이 없으므로 None

total = plus(2, 3)
print(total * 10)
````

````python
# 실습 2-3
def is_high(value, threshold=100):   # threshold의 기본값은 100
    return value >= threshold

print(is_high(120))
print(is_high(120, 150))
````

````python
# 실습 2-4
def to_price(value, rate=0.1, unit="원"):
    return f"{int(value * (1 + rate)):,}{unit}"

print(to_price(10000))
print(to_price(10000, unit="KRW"))     # rate는 기본값 그대로
print(to_price(10000, rate=0.05))      # unit은 기본값 그대로
print(to_price(unit="$", value=10))    # 이름을 적으면 순서는 상관없다
````

````python
# 실습 2-5
def get_summary(values):
    count = len(values)
    total = sum(values)
    return count, total, total / count   # 튜플 하나로 돌려준다

sales = [120000, 150000, 90000, 285000]
n, amount, avg = get_summary(sales)        # 언패킹
print(f"건수 {n}건, 합계 {amount:,}원, 평균 {avg:,.0f}원")
````

````python
# 실습 2-6
orders = [("서울", 300), ("부산", 500), ("대전", 200), ("광주", 450)]

print(sorted(orders, key=lambda row: row[1], reverse=True))   # row[1] = 매출
print(max(orders, key=lambda row: row[1]))
````

````python
# 실습 4-1
stock = {"볼트": 120, "너트": 80}

print(f"볼트 재고: {stock.get('볼트', 0)}개")
print(f"와셔 재고: {stock.get('와셔', 0)}개")   # 없는 키는 기본값 0
````

````python
# 실습 4-2
quantities = ["12", "7개", "30", "", "5"]

valid = []
for q in quantities:
    try:
        valid.append(int(q))
    except ValueError:            # int("7개"), int("")는 ValueError
        print(f'"{q}" → 제외')

print(valid)
print(f"합계: {sum(valid)}개")
````

````python
# 실습 4-3
def summarize(values):
    numbers = []
    for v in values:
        try:
            numbers.append(int(v))
        except ValueError:
            print(f'제외: "{v}"')
    if not numbers:               # 빈 리스트면 평균을 계산하지 않는다
        return 0, 0
    return len(numbers), sum(numbers) / len(numbers)

raw = ["120000", "15만", "90000", "", "285000"]
count, avg = summarize(raw)
print(f"유효 {count}건, 평균 {avg:,.0f}원")

count, avg = summarize([])
if count == 0:
    print("매출 데이터가 없다.")
````
