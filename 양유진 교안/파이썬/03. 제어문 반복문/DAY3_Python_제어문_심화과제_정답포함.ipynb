## 문제 6. 오류 고치기: 점수 판정 코드

아래 코드는 점수마다 판정을 출력하고 우수 인원과 평균을 구하려 했지만 결과가 이상하다. 원인 2가지를 찾아 고친다.

- 90 이상: 우수 / 80 이상: 양호 / 그 외: 보충

다 고쳤으면 `scores = []`로도 실행해 본다. `ZeroDivisionError`가 나지 않고 `점수 데이터가 없다.`만 출력되도록 한 번 더 고친다. 실행 결과는 원래 점수로 실행했을 때와 빈 리스트로 실행했을 때를 이어서 보여 준다.

**실행 결과**

```text
72점: 보충
95점: 우수
88점: 양호
64점: 보충
91점: 우수
우수 2명
평균 82.0점
점수 데이터가 없다.
```

공식 문서: [if 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#if-statements) · [for 문](https://docs.python.org/ko/3.13/tutorial/controlflow.html#for-statements) · [ZeroDivisionError](https://docs.python.org/ko/3.13/library/exceptions.html#ZeroDivisionError)

````python
scores = [72, 95, 88, 64, 91]

for score in scores:
    excellent = 0
    if score >= 80:
        print(f"{score}점: 양호")
    elif score >= 90:
        print(f"{score}점: 우수")
        excellent += 1
    else:
        print(f"{score}점: 보충")

print(f"우수 {excellent}명")
print(f"평균 {sum(scores) / len(scores):.1f}점")
````

## 문제 7. 오류 고치기: 끝나지 않는 반복

[참고] 아래 코드는 1부터 5까지 더하려고 했지만 **실행하면 끝나지 않는다.** 실행하지 말고 코드를 읽어 원인을 찾아 고친 뒤 실행한다. (실수로 실행했다면 Colab 셀 왼쪽의 정지 버튼으로 멈춘다.)

````python
count = 1
total = 0

while count <= 5:
    total += count

print(f"1부터 5까지의 합: {total}")
````

# 정답 확인

먼저 스스로 풀고 비교한다.

### 문제 1. 센서 로그 정리하기 정답

````python
# 문제 1
log = ["78.4", "", "81.2", "err", "76.9", "85.5", "", "79.3"]

empty = 0
error = 0
warning = 0
normal = []

for value in log:
    if not value:                 # 빈 문자열은 거짓
        empty += 1
        continue
    if value == "err":
        error += 1
        continue
    temp = float(value)
    if temp >= 80:
        print(f"경고: {temp}도")
        warning += 1
    else:
        normal.append(temp)

print(f"빈값 {empty}건, 오류 {error}건, 경고 {warning}건")
print(f"정상 값: {normal}")

if normal:                        # 값이 있을 때만 평균 계산
    print(f"정상 평균: {sum(normal) / len(normal):.1f}도")
else:
    print("정상 값이 없다.")
````

### 문제 2. 고객 등급별 인원과 매출 비중 정답

````python
# 문제 2
purchases = [320000, 85000, 150000, 410000, 99000, 175000, 260000, 60000]

vip = 0
good = 0
normal = 0
vip_total = 0
total = 0

for amount in purchases:
    total += amount
    if amount >= 300000:
        vip += 1
        vip_total += amount
    elif amount >= 150000:
        good += 1
    else:
        normal += 1

print(f"VIP {vip}명, 우수 {good}명, 일반 {normal}명")
print(f"VIP 매출 비중: {vip_total / total * 100:.1f}%")
````

### 문제 3. 재고 출고 시뮬레이션 정답

````python
# 문제 3
stock = 95
daily_out = 18
safety = 30

day = 0
while stock >= daily_out:
    stock -= daily_out
    day += 1
    print(f"{day}일째 출고 후 재고: {stock}개")
    if stock < safety:
        print(f"{day}일째 재고 {stock}개 → 발주 필요")
        break
````

### 문제 4. 연속 불량이 나오면 검사 중단하기 정답

````python
# 문제 4
results = ["양품", "불량", "양품", "양품", "불량", "불량", "양품", "불량"]

number = 0
streak = 0
defects = 0

for result in results:
    number += 1
    if result == "불량":
        streak += 1
        defects += 1
    else:
        streak = 0                # 양품이 나오면 연속 기록을 되돌린다
    if streak == 2:
        print(f"{number}번째 제품에서 연속 불량 2회 발생, 검사 중단")
        break

print(f"검사한 제품 수: {number}개, 발견한 불량: {defects}개")
````

### 문제 5. 종합 문제: 일별 매출 분석 정답

````python
# 문제 5
daily_sales = [412000, 389000, 455000, 301000, 478000, 396000, 432000]
target = 400000

if daily_sales:
    achieved = []
    for value in daily_sales:
        if value >= target:
            achieved.append(value)
    print(f"목표 달성: {len(achieved)}일 {achieved}")

    best = daily_sales[0]
    best_day = 1
    day = 0
    for value in daily_sales:
        day += 1
        if value > best:
            best = value
            best_day = day
    print(f"최고 매출: {best_day}일차 {best:,}원")

    average = sum(daily_sales) / len(daily_sales)
    if average >= 430000:
        grade = "우수"
    elif average >= 380000:
        grade = "보통"
    else:
        grade = "개선 필요"
    print(f"평균 매출: {average:,.0f}원 → {grade}")
else:
    print("매출 데이터가 없다.")
````

### 문제 6. 오류 고치기: 점수 판정 코드 정답

````python
# 문제 6
scores = [72, 95, 88, 64, 91]

if scores:                        # ③ 빈 리스트면 평균을 계산하지 않는다
    excellent = 0                 # ① 누적 변수는 반복문 밖에서 시작
    for score in scores:
        if score >= 90:           # ② 더 엄격한 조건을 먼저 검사
            print(f"{score}점: 우수")
            excellent += 1
        elif score >= 80:
            print(f"{score}점: 양호")
        else:
            print(f"{score}점: 보충")
    print(f"우수 {excellent}명")
    print(f"평균 {sum(scores) / len(scores):.1f}점")
else:
    print("점수 데이터가 없다.")

scores = []

if scores:
    excellent = 0
    for score in scores:
        if score >= 90:
            print(f"{score}점: 우수")
            excellent += 1
        elif score >= 80:
            print(f"{score}점: 양호")
        else:
            print(f"{score}점: 보충")
    print(f"우수 {excellent}명")
    print(f"평균 {sum(scores) / len(scores):.1f}점")
else:
    print("점수 데이터가 없다.")
````

### 문제 7. 오류 고치기: 끝나지 않는 반복 정답

````python
# 문제 7
count = 1
total = 0

while count <= 5:
    total += count
    count += 1                    # 조건에 영향을 주는 값을 바꿔야 반복이 끝난다

print(f"1부터 5까지의 합: {total}")
````
