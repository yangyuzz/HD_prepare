# 1. 조건문
📖 교재 4~14쪽

#### ▶ 따라 실행하기

````python
sales = 120000
target = 100000

if sales >= 200000:
    grade = "A등급 (우수)"
elif sales >= target:
    grade = "B등급 (보통)"
else:
    grade = "C등급 (개선 필요)"

print(f"현재 매출: {sales:,}원")
print(f"매장 평가: {grade}")

if sales >= target:
    print(f"초과 달성 금액: {sales - target:,}원")   # 조건이 참일 때만 실행
print("--- 매출 평가 처리 종료 ---")                  # 항상 실행
````

### ✏️ 실습 1-1 · if

defect_rate = 0.05일 때 0.03 이상이면 "재검사 필요"를 출력하는 if 문을 작성한다. 이어서 defect_rate를 0.01로 바꿔 같은 if 문을 실행하면 아무것도 출력되지 않는다.

**실행 결과**
```text
재검사 필요
```

````python
# [실습 1-1] if
# 1. defect_rate에 0.05를 저장한다.
# 2. 0.03 이상이면 "재검사 필요"를 출력한다.
# 3. defect_rate를 0.01로 바꾸고 같은 if 문을 다시 작성한다.

defect_rate = 0.05
if defect_rate >= 0.03:
    print("재검사 필요")

defect_rate = 0.01
if defect_rate >= 0.03:
    print("재검사 필요")   # 조건이 거짓이라 출력되지 않는다
````

### ✏️ 실습 1-2 · if-else

stock = 8, safety_stock = 10일 때 재고가 안전 재고보다 적으면 "발주 필요", 아니면 "재고 충분"을 출력한다.

**실행 결과**
```text
발주 필요
```

````python
# [실습 1-2] if-else
# 1. stock과 safety_stock을 저장한다.
# 2. if-else로 결과를 출력한다.

stock = 8
safety_stock = 10

if stock < safety_stock:
    print("발주 필요")
else:
    print("재고 충분")
````

### ✏️ 실습 1-3 · if-elif-else

온도 temperature가 78.5일 때 80 이상이면 "경고", 70 이상이면 "주의", 그 외에는 "정상"을 status에 저장하고, 조건문 밖에서 온도와 상태를 출력한다.

**실행 결과**
```text
온도 78.5도: 주의
```

````python
# [실습 1-3] if-elif-else
# 1. temperature에 78.5를 저장한다.
# 2. if-elif-else로 status를 정한다.
# 3. 조건문 밖에서 실행 결과 형식으로 출력한다.

temperature = 78.5

if temperature >= 80:
    status = "경고"
elif temperature >= 70:
    status = "주의"
else:
    status = "정상"

print(f"온도 {temperature}도: {status}")
````

### ✏️ 실습 1-4 · 고객 등급 분류

구매 금액 purchase_amount가 165000원일 때 200,000원 이상이면 VIP, 100,000원 이상이면 우수, 그 외는 일반을 customer_grade에 저장하고, 구매 금액에 천 단위 쉼표를 붙여 출력한다.

**실행 결과**
```text
구매 금액: 165,000원 ➔ 고객 등급: 우수
```

````python
# [실습 1-4] 고객 등급 분류
# 1. purchase_amount를 저장한다.
# 2. if-elif-else로 customer_grade를 정한다.
# 3. f-string으로 출력한다.

purchase_amount = 165000

if purchase_amount >= 200000:
    customer_grade = "VIP"
elif purchase_amount >= 100000:
    customer_grade = "우수"
else:
    customer_grade = "일반"

print(f"구매 금액: {purchase_amount:,}원 ➔ 고객 등급: {customer_grade}")
````

### ✏️ 실습 1-5 · 조건문과 논리 연산자

주문 금액 amount = 320000, 회원 여부 is_member = False이다.

- 주문 금액이 300000원 이상이거나 회원이면 "무료 배송", 아니면 "배송비 3,000원"
- 주문 금액이 500000원 이상이고 회원이면 "VIP 할인 적용", 아니면 "일반 결제"

**실행 결과**
```text
무료 배송
일반 결제
```

````python
# [실습 1-5] 조건문과 논리 연산자
# 1. amount와 is_member를 저장한다.
# 2. or 조건으로 배송비를 판단해 출력한다.
# 3. and 조건으로 할인 여부를 판단해 출력한다.

amount = 320000
is_member = False

if amount >= 300000 or is_member:
    print("무료 배송")
else:
    print("배송비 3,000원")

if amount >= 500000 and is_member:
    print("VIP 할인 적용")
else:
    print("일반 결제")
````

# 2. 반복문
📖 교재 15~26쪽

#### ▶ 따라 실행하기

````python
sales_list = [120000, 180000, 90000, 210000, 150000]

total = 0
high_count = 0
for sales in sales_list:
    total += sales              # 누적 합계
    if sales >= 150000:
        high_count += 1         # 조건에 맞을 때만 개수 증가
        print(f"고매출: {sales:,}원")

print(f"합계: {total:,}원, 고매출 {high_count}건")
````

### ✏️ 실습 2-1 · for 문으로 값 꺼내기

설비 목록 machines = ["M01", "M02", "M03"]을 for 문으로 하나씩 꺼내 출력한다.

**실행 결과**

````python
# [실습 2-1] for 문으로 값 꺼내기
# 1. machines를 저장한다.
# 2. for 문으로 하나씩 출력한다.

machines = ["M01", "M02", "M03"]

for machine in machines:
    print(machine)
````

### ✏️ 실습 2-2 · 누적 합계

sales = [100000, 150000, 200000]을 for 문으로 하나씩 더해 전체 매출을 출력한다. 누적 합계 변수 total은 0으로 시작한다.

**실행 결과**
```text
전체 매출: 450000원
```

````python
# [실습 2-2] 누적 합계
# 1. sales와 total = 0을 준비한다.
# 2. for 문으로 total에 더한다.
# 3. 반복이 끝난 뒤 출력한다.

sales = [100000, 150000, 200000]
total = 0

for value in sales:
    total += value

print(f"전체 매출: {total}원")
````

### ✏️ 실습 2-3 · for + if: 조건에 맞는 값 출력

점수 목록 scores = [85, 60, 90, 70, 95]에서 80점 이상인 점수만 출력한다.

**실행 결과**
```text
85
90
95
```

````python
# [실습 2-3] for + if: 조건에 맞는 값 출력
# 1. scores를 저장한다.
# 2. for 문 안에서 if로 80 이상만 출력한다.

scores = [85, 60, 90, 70, 95]

for score in scores:
    if score >= 80:
        print(score)
````

### ✏️ 실습 2-4 · for + if: 개수 세기

2-3의 scores에서 80점 이상인 점수의 개수를 세어 출력한다. 개수 변수 count는 0으로 시작한다.

**실행 결과**
```text
80점 이상: 3개
```

````python
# [실습 2-4] for + if: 개수 세기
# 1. count = 0을 준비한다.
# 2. 조건에 맞을 때만 count를 1씩 늘린다.
# 3. 반복이 끝난 뒤 출력한다.

count = 0

for score in scores:
    if score >= 80:
        count += 1

print(f"80점 이상: {count}개")
````

### ✏️ 실습 2-5 · for + if: 조건에 맞는 값으로 리스트 만들기

numbers = [10, 25, 30, 15, 40]에서 20 이상인 숫자만 골라 result 리스트에 추가하고 출력한다. result는 빈 리스트로 시작한다.

**실행 결과**
```text
[25, 30, 40]
```

````python
# [실습 2-5] for + if: 조건에 맞는 값으로 리스트 만들기
# 1. numbers와 빈 리스트 result를 준비한다.
# 2. 20 이상이면 append()로 추가한다.
# 3. result를 출력한다.

numbers = [10, 25, 30, 15, 40]
result = []

for number in numbers:
    if number >= 20:
        result.append(number)

print(result)
````

### ✏️ 실습 2-6 · 종합: 온도 경고

temperatures = [72.1, 81.5, 68.9, 79.8, 85.2]에서 80도 이상이면 "경고: 81.5도" 형식으로 출력하고, 경고 건수는 warning_count에, 80도 미만 온도는 normal 리스트에 모아 출력한다.

**실행 결과**
```text
경고: 81.5도
경고: 85.2도
경고 건수: 2건
정상 온도: [72.1, 68.9, 79.8]
```

````python
# [실습 2-6] 종합: 온도 경고
# 1. temperatures, warning_count = 0, normal = []을 준비한다.
# 2. 80 이상이면 경고를 출력하고 개수를 센다.
# 3. 아니면 normal에 추가한다.
# 4. 경고 건수와 정상 온도를 출력한다.

temperatures = [72.1, 81.5, 68.9, 79.8, 85.2]
warning_count = 0
normal = []

for t in temperatures:
    if t >= 80:
        print(f"경고: {t}도")
        warning_count += 1
    else:
        normal.append(t)

print(f"경고 건수: {warning_count}건")
print(f"정상 온도: {normal}")
````

### ✏️ 실습 2-7 · 문자열 리스트를 숫자 리스트로

sales_text = "85000,120000,97000"을 split(",")으로 나눈 뒤, for 문으로 하나씩 int()로 바꿔 sales_numbers에 모으고 합계를 출력한다.

**실행 결과**
```text
[85000, 120000, 97000]
매출 합계: 302,000원
```

````python
# [실습 2-7] 문자열 리스트를 숫자 리스트로
# 1. sales_text를 split()으로 나눠 parts에 저장한다.
# 2. 빈 리스트 sales_numbers를 만든다.
# 3. for 문으로 int()로 바꿔 append()한다.
# 4. sales_numbers와 합계를 출력한다.

sales_text = "85000,120000,97000"
parts = sales_text.split(",")

sales_numbers = []
for value in parts:
    sales_numbers.append(int(value))

print(sales_numbers)
print(f"매출 합계: {sum(sales_numbers):,}원")
````

### ✏️ 실습 2-8 · continue: 건너뛰기

일별 생산량 production = [120, 135, 0, 142, 0, 130]에서 0(휴무일)은 continue로 건너뛰고, 가동일 수와 생산량 합계를 출력한다.

**실행 결과**
```text
가동일 4일, 합계 527
```

````python
# [실습 2-8] continue: 건너뛰기
# 1. production, days = 0, total = 0을 준비한다.
# 2. 0이면 continue로 건너뛴다.
# 3. 아니면 days와 total을 늘린다.
# 4. 결과를 출력한다.

production = [120, 135, 0, 142, 0, 130]
days = 0
total = 0

for amount in production:
    if amount == 0:
        continue
    days += 1
    total += amount

print(f"가동일 {days}일, 합계 {total}")
````

### ✏️ 실습 2-9 · break: 반복 멈추기

temperatures = [72.1, 75.4, 83.0, 79.8, 85.2, 92.0, 76.5]를 순서대로 확인한다.

- 80도 미만: "정상"을 출력하고 continue로 다음 온도를 본다
- 90도 이상: "위험 경보, 점검 중단"을 출력하고 break로 끝낸다
- 그 외(80도 이상): "경고"를 출력한다

**실행 결과**
```text
72.1도: 정상
75.4도: 정상
83.0도: 경고
79.8도: 정상
85.2도: 경고
92.0도: 위험 경보, 점검 중단
```

````python
# [실습 2-9] break: 반복 멈추기
# 1. temperatures를 저장한다.
# 2. 80 미만이면 정상 출력 후 continue한다.
# 3. 90 이상이면 위험 경보 출력 후 break한다.
# 4. 나머지는 경고를 출력한다.

temperatures = [72.1, 75.4, 83.0, 79.8, 85.2, 92.0, 76.5]

for t in temperatures:
    if t < 80:
        print(f"{t}도: 정상")
        continue
    if t >= 90:
        print(f"{t}도: 위험 경보, 점검 중단")
        break
    print(f"{t}도: 경고")
````

### ✏️ 실습 3-7 · ZeroDivisionError: 빈 리스트 평균

리스트가 비어 있으면 "데이터 없음"을 출력하고, 아니면 평균을 출력하도록 조건문을 추가한다.
