# 정답 확인

### 1. 튜플 정답

````python
# 문제 1-1
equipment = ("CNC-03", "2공장", 2019)

print(equipment)
print(type(equipment))
print(equipment[0])
print(equipment[-1])
print(len(equipment))

# equipment[2] = 2020   # TypeError: 튜플은 만든 뒤 값을 바꿀 수 없다
````

````python
# 문제 1-2
branch_data = ("강남점", 720000, 48)

branch, sales, quantity = branch_data   # 변수 개수와 값 개수가 같아야 한다

print(f"지점명: {branch}")
print(f"매출: {sales:,}원")
print(f"판매수량: {quantity}개")
print(f"1개당 평균 금액: {sales / quantity:,.0f}원")
````

````python
# 문제 1-3
checks = ("정상", "정상", "경고", "정상", "이상", "경고", "정상", "정상")

warn_count = checks.count("경고")
error_count = checks.count("이상")
first_warn = checks.index("경고")       # 처음 나타나는 위치
normal_rate = checks.count("정상") / len(checks)

print(f"경고: {warn_count}회, 이상: {error_count}회")
print(f"첫 경고 위치(인덱스): {first_warn}")
print(f"정상 비율: {normal_rate:.1%}")
````

### 2. 딕셔너리 정답

````python
# 문제 2-1
stock = {"볼트": 1200, "너트": 950, "와셔": 2300}

print(stock["너트"])

stock["스프링"] = 600   # 없는 키에 대입하면 추가된다
stock["볼트"] = 1050    # 있는 키에 대입하면 수정된다

print(stock)
````

````python
# 문제 2-2
region_sales = {"서울": 350000, "부산": 280000, "대구": 210000}

print(region_sales.get("대전", 0))   # 키가 없으면 기본값 0
print(region_sales.get("부산", 0))   # 키가 있으면 저장된 값
print("대전" in region_sales)        # 딕셔너리의 in 은 키를 확인한다

# region_sales["대전"]   # KeyError: 없는 키를 대괄호로 조회하면 오류가 난다
````

````python
# 문제 2-3
line_output = {"A라인": 4120, "B라인": 3870, "C라인": 4450}

print(list(line_output.keys()))
print(list(line_output.values()))
print(list(line_output.items()))   # 각 쌍은 튜플이다

# 방법 1: 세 값을 직접 더한다
total = line_output["A라인"] + line_output["B라인"] + line_output["C라인"]
print(f"라인 수: {len(line_output)}개, 총생산량: {total:,}개")

# 방법 2: sum()으로 값 전체를 한 번에 더한다 (값 뷰를 바로 넣어도 된다)
total_sum = sum(line_output.values())
print(f"sum()으로 계산한 총생산량: {total_sum:,}개")
````

````python
# 문제 2-4
price = {"모니터": 280000, "키보드": 45000, "마우스": 23000}

price.update({"키보드": 42000, "헤드셋": 89000})   # 있으면 수정, 없으면 추가
print(price)

removed = price.pop("마우스")       # 삭제한 값을 돌려준다
print(f"삭제한 마우스 단가: {removed:,}원")

print(price.pop("스피커", 0))       # 없는 키는 기본값을 돌려준다

del price["모니터"]
print(price)
````

### 3. 집합 정답

````python
# 문제 3-1
visit_log = ["울산", "창원", "울산", "구미", "창원", "울산", "포항"]

regions = set(visit_log)
print(len(regions))

regions.add("거제")
regions.discard("구미")
regions.discard("여수")   # 없는 값이어도 오류가 나지 않는다 (remove() 는 KeyError)

print("구미" in regions)
print(sorted(regions))    # 집합은 순서가 없으므로 정렬해서 확인한다
````

````python
# 문제 3-2
january = {"김영희", "이철수", "박민수", "최지우"}
february = {"이철수", "최지우", "정수민"}

print(sorted(january & february))   # 교집합: 두 달 모두
print(sorted(january - february))   # 차집합: 1월에만
print(len(january | february))      # 합집합: 한 번이라도
````
