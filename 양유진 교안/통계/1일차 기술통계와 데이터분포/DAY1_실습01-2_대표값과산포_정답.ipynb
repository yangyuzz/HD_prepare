### 문제 6. ddof=0과 ddof=1 비교
`PPT p.60, 59`

데이터 `[2, 4, 6]`의 평균과 편차를 구하고, ddof=0과 ddof=1의 분산·표준편차를 소수 넷째 자리까지 비교하시오.

**확인** : ddof=0 → 2.6667 / 1.6330, ddof=1 → 4.0000 / 2.0000

```python
example = pd.Series([2, 4, 6])
print("평균:", example.mean(), "/ 편차:", (example - example.mean()).tolist())
for d in [0, 1]:
    print(f"ddof={d} → 분산 {example.var(ddof=d):.4f}, 표준편차 {example.std(ddof=d):.4f}")
```

**확인** : 99.38 (Nm)², 9.97 Nm

```python
print("분산:", round(torque.var(ddof=1), 2), "(Nm)²")
print("표준편차:", round(torque.std(ddof=1), 2), "Nm")
```

```python
# [데이터 준비] 문제 8에 사용한다. 그대로 실행한다
process_a = pd.Series([76, 80, 84, 80, 80], name="공정 A")
process_b = pd.Series([18, 20, 22, 20, 20], name="공정 B")
```

### 문제 8. 변동계수(CV) 비교
`PPT p.63`

공정 A와 B의 평균, 표준편차(ddof=1), CV(%)를 한 표로 모아 소수 둘째 자리까지 출력하고, 어느 공정의 상대적 변동이 더 큰지 쓰시오.

**확인** : A : 80, 2.83, 3.54% / B : 20, 1.41, 7.07%

```python
rows = {}
for s in [process_a, process_b]:
    m, sd = s.mean(), s.std(ddof=1)
    rows[s.name] = {"평균": m, "표준편차": sd, "CV(%)": sd / m * 100}
display(pd.DataFrame(rows).T.round(2))
# 표준편차는 A가 크지만, CV는 B가 커서 평균 대비 변동은 B가 더 크다
```

## 종합 실습 : 두 라인의 사이클타임

사이클타임(cycle time) : 제품 한 개를 생산하는 데 걸린 시간(초)

| 열 | 뜻 |
|---|---|
| 라인 | 생산 라인 (A·B) |
| 사이클타임 | 제품 한 개 생산에 걸린 시간 (초) |

```python
# [데이터 준비] 그대로 실행한다. np.nan : 기록이 빠진 값(결측). 라인 B에 하나 있다
cycle = pd.DataFrame({
    "라인":       ["A", "A", "A", "A", "A", "A", "B", "B", "B", "B", "B", "B"],
    "사이클타임": [58,  61,  60,  59,  62,  180, 118, 122, np.nan, 121, 117, 122],
})
line_a = cycle[cycle["라인"] == "A"]["사이클타임"]
line_b = cycle[cycle["라인"] == "B"]["사이클타임"]
cycle
```

### 문제 종합 1. 행 수와 유효값 수
`PPT p.30`

전체 행 수와 사이클타임의 유효값 수는 각각 몇 개인가? 두 값이 다른 이유는?

**확인** : 12, 11

```python
print("전체 행:", len(cycle), "/ 유효값:", cycle["사이클타임"].count())
# 라인 B에 결측(NaN)이 1개 있어 유효값이 1개 적다
```

### 문제 종합 2. 라인 A의 평균과 중앙값
`PPT p.41`

라인 A의 평균과 중앙값을 구하고, 두 값이 크게 다른 이유를 설명하시오.

**확인** : 80.0, 60.5

```python
print("평균:", round(line_a.mean(), 2), "/ 중앙값:", line_a.median())
# 180초 하나가 평균만 크게 끌어올렸다
```

### 문제 종합 3. 범위와 IQR
`PPT p.47, 49`

라인 A의 범위와 IQR을 구하고, 어느 쪽이 큰 값 180의 영향을 더 크게 받는지 설명하시오.

**확인** : 범위 122, IQR 2.5

```python
q1, q3 = line_a.quantile([0.25, 0.75])
print("범위:", line_a.max() - line_a.min(), "/ IQR:", q3 - q1)
# 범위는 180을 그대로 반영해 커지고, IQR은 가운데 50%만 봐서 거의 영향이 없다
```

### 문제 종합 4. 표준편차 비교
`PPT p.60`

확인해 보니 180초는 설비 정지로 생긴 기록이었다. 이 값을 뺀 라인 A(`line_a_normal`)와 라인 B의 표준편차를 구하시오.
(힌트 : `line_a_normal = line_a[line_a != 180]`)

**확인** : A 1.58, B 2.35

```python
line_a_normal = line_a[line_a != 180]
print("A(정상):", round(line_a_normal.std(), 2), "/ B:", round(line_b.std(), 2))
```

### 문제 종합 5. CV로 다시 비교
`PPT p.63`

라인 A(정상)와 라인 B의 CV(%)를 구하고, 종합 4의 표준편차 비교와 결론이 어떻게 달라지는지 설명하시오.

**확인** : A 2.64%, B 1.95%

```python
line_a_normal = line_a[line_a != 180]
for name, s in [("A(정상)", line_a_normal), ("B", line_b)]:
    print(name, "CV(%):", round(s.std() / s.mean() * 100, 2))
# 표준편차는 B가 크지만, 평균 대비로 보면 A(약 2.6%)가 B(약 2.0%)보다 더 흔들린다
```

### 문제 종합 6. describe() 읽기
`PPT p.65`

라인 B의 `describe()` 결과에서 count, std, 50%는 각각 얼마이며 무엇을 뜻하는가?

**확인** : count 5, std 2.35, 50% 121

```python
display(line_b.describe())
# count 5 : 결측을 뺀 유효값 수 / std 2.35 : 표준편차(ddof=1) / 50% 121 : 중앙값
```
