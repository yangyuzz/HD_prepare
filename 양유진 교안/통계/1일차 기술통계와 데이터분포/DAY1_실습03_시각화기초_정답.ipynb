```python
import os

from google.colab import drive

drive.mount("/content/drive")
data_folder = "/content/drive/MyDrive/HD-AI/data"
csv_path = os.path.join(data_folder, "ai4i2020.csv")

df = pd.read_csv(csv_path)   # 원본
data = df.copy()             # 실습용 사본
```

```python
# [데이터 준비] 자주 쓰는 열을 이름에 담아 둔다
torque = data["Torque [Nm]"]
speed = data["Rotational speed [rpm]"]
```

### 문제 1. 세 도구로 같은 히스토그램
`PPT p.106, 105`

토크 히스토그램(bins=30)을 Pandas, Matplotlib, Seaborn으로 각각 그리고 제목에 도구 이름을 적으시오.

```python
torque.hist(bins=30); plt.title("Pandas : torque.hist()"); plt.show()
plt.hist(torque, bins=30); plt.title("Matplotlib : plt.hist()"); plt.show()
sns.histplot(torque, bins=30); plt.title("Seaborn : sns.histplot()"); plt.show()
```

### 문제 2. 기준선과 범례
`PPT p.109`

그림 크기 (8, 4)로 토크 히스토그램(bins=30, 하늘색 막대, 흰 테두리)을 그리고 평균(빨간 점선)과 중앙값(초록 짧은 점선) 세로선, 범례, 제목·축 이름을 넣으시오.

```python
plt.figure(figsize=(8, 4))
plt.hist(torque, bins=30, color="skyblue", edgecolor="white")
plt.axvline(torque.mean(), color="red", linestyle="--", label="평균")
plt.axvline(torque.median(), color="green", linestyle=":", label="중앙값")
plt.title("토크 분포와 평균·중앙값"); plt.xlabel("토크(Nm)"); plt.ylabel("기록 수")
plt.legend(); plt.show()
```

### 문제 3. 범주별 개수 막대그래프
`PPT p.110`

Type별 기록 수를 먼저 계산한 뒤 막대그래프로 그리시오. 가로축 글자는 세워서 표시한다.

```python
type_cnt = data["Type"].value_counts()
type_cnt.plot.bar(color="steelblue")
plt.title("Type별 기록 수"); plt.xlabel("Type"); plt.ylabel("기록 수"); plt.xticks(rotation=0)
plt.show()
```

### 문제 4. 범주별 고장률 막대그래프
`PPT p.110`

`groupby`로 Type별 고장률(%)을 계산해 막대그래프로 그리시오.

**확인** : H 2.09, L 3.92, M 2.77

```python
fail_rate = data.groupby("Type")["Machine failure"].mean() * 100
print(fail_rate.round(2))
fail_rate.plot.bar(color="tomato")
plt.title("Type별 고장률"); plt.xlabel("Type"); plt.ylabel("고장률(%)"); plt.xticks(rotation=0)
plt.show()
```

### 문제 5. 구간 경계와 개수 돌려받기
`PPT p.111`

`scores = [10, 12, 18, 19, 20, 21, 26, 36, 45, 50]`로 bins=5 히스토그램을 그리고, 돌려받은 구간 경계와 구간별 개수를 출력하시오.

**확인** : 경계 10, 18, 26, 34, 42, 50 / 개수 2, 4, 1, 1, 2

```python
scores = [10, 12, 18, 19, 20, 21, 26, 36, 45, 50]
counts, edges, patches = plt.hist(scores, bins=5, edgecolor="black")
plt.title("bins=5 히스토그램"); plt.xlabel("값"); plt.ylabel("개수"); plt.show()
print("구간 경계:", edges)
print("구간별 개수:", counts)
```

### 문제 6. 이상값 후보 점 숨기기
`PPT p.112`

회전속도 박스플롯을 기본(평균 표시)과 `showfliers=False` 두 가지로 그리고, 세로축 범위가 어떻게 달라지는지 주석으로 적으시오.

```python
speed.plot.box(showmeans=True); plt.title("회전속도 박스플롯 (기본)"); plt.ylabel("회전속도(rpm)"); plt.show()
speed.plot.box(showmeans=True, showfliers=False); plt.title("회전속도 박스플롯 (이상값 후보 숨김)"); plt.ylabel("회전속도(rpm)"); plt.show()
# 점이 사라져 세로축 범위가 좁아지고 상자가 크게 보인다 (데이터는 그대로)
```

### 문제 7. 집단별 박스플롯
`PPT p.113`

Seaborn으로 Type별 토크 박스플롯을 L, M, H 순서로 그리시오.

```python
sns.boxplot(data=data, x="Type", y="Torque [Nm]", order=["L", "M", "H"])
plt.title("Type별 토크"); plt.xlabel("Type"); plt.ylabel("토크(Nm)")
plt.show()
```

### 문제 8. 두 그래프 나란히
`PPT p.114`

`plt.subplots()`로 가로 2칸을 만들고 왼쪽에 토크 히스토그램, 오른쪽에 토크 박스플롯을 그리시오. 칸마다 제목·축 이름을 넣는다.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
torque.hist(bins=30, ax=axes[0])
axes[0].set_title("토크 히스토그램"); axes[0].set_xlabel("토크(Nm)"); axes[0].set_ylabel("기록 수")
torque.plot.box(showmeans=True, ax=axes[1])
axes[1].set_title("토크 박스플롯"); axes[1].set_ylabel("토크(Nm)")
plt.tight_layout(); plt.show()
```

### 문제 10. 공정 온도 히스토그램 [Matplotlib]
`PPT p.109`

`Process temperature [K]`의 히스토그램을 bins=30으로 그리고, 평균 위치에 빨간 점선을 그으시오.

```python
proc = data["Process temperature [K]"]
plt.hist(proc, bins=30, edgecolor="white")
plt.axvline(proc.mean(), color="red", linestyle="--", label=f"평균 {proc.mean():.2f}K")
plt.title("공정 온도 분포"); plt.xlabel("공정 온도(K)"); plt.ylabel("기록 수"); plt.legend()
plt.show()
```

### 문제 11. Type별 공구 마모 시간 [Seaborn]
`PPT p.113`

Type별 `Tool wear [min]` 박스플롯을 L, M, H 순서로 그리고, 세 Type의 차이를 주석으로 설명하시오.

```python
sns.boxplot(data=data, x="Type", y="Tool wear [min]", order=["L", "M", "H"])
plt.title("Type별 공구 마모 시간"); plt.ylabel("공구 마모(min)")
plt.show()
print(data.groupby("Type")["Tool wear [min]"].median())
# 세 Type의 중앙값(약 106~109분)과 IQR이 거의 같다 → Type별 차이는 작다
```
