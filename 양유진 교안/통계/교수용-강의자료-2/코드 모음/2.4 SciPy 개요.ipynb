```python
# 2.4 SciPy 개요
# 2.4.1 SciPy 가져오기
# SciPy 가져와서 확률분포 살펴보기
#
import numpy as np

from scipy import stats

# 정규 분포의 평균과 표준 편차 설정
mean = 0
std_dev = 1

# 정규 분포 객체 생성
norm_dist = stats.norm(loc=mean, scale=std_dev)

# 확률 밀도 함수(PDF) 계산
x = np.linspace(-5, 5, 100)  # X 값 범위
pdf = norm_dist.pdf(x)  # PDF 값 계산

# 확률 분포 그래프
import matplotlib.pyplot as plt
plt.plot(x, pdf, label='PDF')
plt.title('Normal Distribution PDF')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```
