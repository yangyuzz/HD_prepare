```python
# 4.6 파이썬 실습
# 4.6.1 함수의 그래프 그리기 및 극한 계산
# 극한 계산
#
import numpy as np
import matplotlib.pyplot as plt
# 함수 정의: f(x) = sin(x) / x
def f(x):
    return np.where(x == 0, 1.0, np.sin(x) / x)  # x가 0일 때 1을 반환하도록 수정
# x값 범위 설정 (-50, 50)
x = np.linspace(-50, 50, 400)
y = f(x)
# 그래프 그리기
plt.plot(x, y, label="f(x) = sin(x)/x")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.title("Graph of f(x) = sin(x)/x")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
# 극한 값 출력
print("The limit of f(x) as x approaches 0 is 1")
```

```python
# 4.6 파이썬 실습
# 4.6.1 함수의 그래프 그리기 및 극한 계산
# 0 주변의 극한값
#
import numpy as np
import matplotlib.pyplot as plt

# x의 범위 설정: 0부터 π/2까지
x = np.linspace(0, np.pi/2, 100)

# y = x와 y = sin(x) 계산
y_linear = x
y_sine = np.sin(x)

# 그래프 그리기
plt.plot(x, y_linear, label='y = x')
plt.plot(x, y_sine, label='y = sin(x)')
plt.title('Graphs of y = x and y = sin(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 4.6 파이썬 실습
# 4.6.2 함수의 연속성과 불연속성 시각화
# 불연속점
#
import numpy as np
import matplotlib.pyplot as plt

# 함수 g(x) 정의
def g(x):
    return np.where(x != 2, x**2, 5)

# x값 범위 설정
x = np.linspace(1, 3, 400)
y = g(x)

# x=2를 제외한 점들만 플롯
x_except_2 = x[x != 2]
y_except_2 = y[x != 2]

# 그래프 그리기
plt.plot(x_except_2, y_except_2, label='g(x)')
plt.scatter([2], [5], color='red', s=50, zorder=5)  # x = 2에서의 함수값 강조
plt.scatter([2], [4], edgecolor='black', facecolor='white', s=50, zorder=5, label='(2, 4)')  # (2, 4) 점을 속이 빈 동그라미로 표시
plt.title('Graph of g(x)')
plt.xlabel('x')
plt.ylabel('g(x)')
plt.legend()
plt.grid(True)
plt.show()

# x = 2에서의 좌극한과 우극한 계산
left_limit = 2**2
right_limit = 2**2
function_value = 5

# 연속성 검증
print(f"Left limit at x = 2: {left_limit}")
print(f"Right limit at x = 2: {right_limit}")
print(f"Function value at x = 2: {function_value}")

if left_limit == right_limit == function_value:
    print("The function is continuous at x = 2.")
else:
    print("The function is discontinuous at x = 2.")
```

```python
# 4.6 파이썬 실습
# 4.6.3 미분의 기본 정의와 규칙
# 도함수
#
import numpy as np
import matplotlib.pyplot as plt
# 함수 f(x) 정의
def f(x):
    return x**2 + 2*x + 1
# 수치적 미분을 위한 함수
def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)
# x 값 범위 설정
x = np.linspace(-10, 10, 400)
# 원래 함수와 도함수 값 계산
y = f(x)
y_derivative = numerical_derivative(f, x)
# 그래프 그리기
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(x, y, label='f(x) = x^2 + 2x + 1')
plt.title('Function f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.subplot(1, 2, 2)
plt.plot(x, y_derivative, label="f'(x)", color='orange')
plt.title("Derivative of f(x)")
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 4.6 파이썬 실습
# 4.6.4 적분의 응용
# 원의 면적
#
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
# 원의 반지름 설정
r = 3
# 원의 방정식에 따른 y값 계산 함수
def circle_upper(x):
    return np.sqrt(r**2 - x**2)
def circle_lower(x):
    return -np.sqrt(r**2 - x**2)
# 적분을 통한 면적 계산
area, _ = quad(circle_upper, -r, r)
area = 2 * area  # 상단과 하단 면적의 합
# 원 그리기
x = np.linspace(-r, r, 1000)
y_upper = circle_upper(x)
y_lower = circle_lower(x)
plt.fill_between(x, y_lower, y_upper, color='lightblue')
plt.title(f"Area of Circle with radius {r}")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axis('equal')
plt.show()
print(f"The area of the circle is approximately {area:.2f}")
```

```python
# 4.6 파이썬 실습
# 4.6.5 편미분과 편도함수
# 다변수 함수의 편미분과 편도함수
#
import sympy as sp

x, y = sp.symbols('x y')
f = x**2 * y + y**3

# x에 대한 편미분
f_px = sp.diff(f, x)

# y에 대한 편미분
f_py = sp.diff(f, y)

print("Partial derivative with respect to x:", f_px)
print("Partial derivative with respect to y:", f_py)
```

```python
# 4.6 파이썬 실습
# 4.6.6 편미분의 기하학적 의미
# 3D 표면 그래프와 등고선 그래프
#
from mpl_toolkits.mplot3d import Axes3D
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = x**2 - y**2
fig = plt.figure(figsize=(12, 6))
# 3D 표면 그래프
ax = fig.add_subplot(1, 2, 1, projection='3d')
ax.plot_surface(x, y, z, cmap='viridis')
ax.set_title('Surface plot')
# 등고선 그래프
ax = fig.add_subplot(1, 2, 2)
contour = ax.contour(x, y, z, 20, cmap='viridis')
fig.colorbar(contour)
ax.set_title('Contour plot')
plt.show()
```

```python
# 4.6 파이썬 실습
# 4.6.6 편미분의 기하학적 의미
# 이중적분 계산과 시각화
#
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import dblquad
# 이중적분할 함수 정의
def integrand(x, y):
    return x**2 + y**2
# 적분 범위 설정
x_lower = 0
x_upper = 1
y_lower = 0
y_upper = 1
# 이중적분 계산
integral, error = dblquad(integrand, x_lower, x_upper, lambda x: y_lower, lambda x: y_upper)
# 함수와 적분 영역 시각화
x = np.linspace(x_lower, x_upper, 30)
y = np.linspace(y_lower, y_upper, 30)
X, Y = np.meshgrid(x, y)
Z = integrand(X, Y)
fig = plt.figure()
ax = plt.axes(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_title('Surface plot of f(x, y) = x^2 + y^2')
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
plt.show()
print(f"The integral value over the square [0,1]x[0,1] is: {integral}")
```

```python
# 4.6 파이썬 실습
# 4.6.7 중적분의 응용
# scipy.integrate.dblquad 함수 사용 원의 면적 구하기
#
import numpy as np
from scipy.integrate import dblquad
# 원의 방정식에 따른 y값 계산 함수
def circle_upper(x):
    return np.sqrt(1 - x**2)  # 원의 상단 부분
def circle_lower(x):
    return -np.sqrt(1 - x**2)  # 원의 하단 부분
# 적분을 통한 면적 계산
area, error = dblquad(lambda y, x: 1, -1, 1, circle_lower, circle_upper)
print(f"The area of the circle is approximately {area:.2f}")
```

```python
# 4.6 파이썬 실습
# 4.6.8 편미분의 응용
# scipy.optimize의 최적화 함수를 사용한 함수 최솟값 구하기
#
from scipy.optimize import minimize
def f(variables):
    x, y = variables
    return (x - 1)**2 + (y - 2)**2
initial_guess = [0, 0]
result = minimize(f, initial_guess)
print("The minimum of the function occurs at:", result.x)
```

```python
# 4.6 파이썬 실습
# 4.6.9 중적분의 기본 개념 및 계산
# scipy.integrate.dblquad 함수를 사용한 이중적분
#
from scipy import integrate

# 이중적분할 함수 정의
def f(x, y):
    return x**2 + y**2

# 이중적분 수행
result, error = integrate.dblquad(f, 0, 1, lambda x: 0, lambda x: 1)

# 결과 출력
print("중적분 값:", result)
```
