```python
# 2.1 파이썬 개요
# 2.1.3 연산자와 수식
# 사칙 연산자를 이용하여 연산과 결과를 변수에 저장하고 출력
#
# 산술 연산자를 이용한 다양한 연산
a = 7
b = 5

c = a + b   # 덧셈
print ( "덧셈:", c)

d = a - b   # 뺄셈
print ( "뺄셈:", d)

e = a * b   # 곱셈
print ( "곱셈:", e)

f = a / b   # 나눗셈(소수점 이하 버리지 않음)
print ( "나눗셈(소수점 이하 버리지 않음):", f)

g = a // b   # 나눗셈(소수점 이하 버림)
print ( "나눗셈(소수점 이하 버림):", g)

h = a % b   # 나눗셈(나머지)
print ( "나눗셈(나머지):", h)

i = a ** b  # 거듭제곱(a의 b승)
print ( "거듭제곱:", i)
```

```python
# 2.1 파이썬 개요
# 2.1.4 변수의 유형과 문자열
# 정수형과 부동소수점형
#
x = 2
y = 1.2
print("x의 유형은", type(x)) # 정수형
print("y의 유형은", type(y)) # 부동소수점 형
z = x + y
print("z = ", z)
print("z의 유형은", type(z)) # 부동소수점 형
# 정수형과 부동소수점 형의 합은 부동소수점 형
i = 1.3e4 # 132 × 104을 나타낸다
print("i= ", i)
print("i의 유형은", type (i))
j = 1e-4 # 1 × 10-4를 나타낸다
print("j = ", j)
print("j의 유형은", type(j))
```

```python
# 2.1 파이썬 개요
# 2.1.10 함수
# 매개변수에 전달하는 함수
#
# 매개변수에 전달하는 함수
def get_min(x, y): # x, y는 매개변수
    if (x < y): # x가 y보다 작은지 비교한다
        return x
    else:
       return y
print(get_min(10, 20)) # 함수가 호출되면 매개변수에 값을 전달
```
