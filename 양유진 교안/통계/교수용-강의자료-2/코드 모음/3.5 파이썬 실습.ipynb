```
# 3.5 파이썬 실습
# 3.5.1 집합 연산 구현
#
# 점심 메뉴
lunch_menu = {"스테이크", "파스타", "샐러드", "스프"}
# 저녁 메뉴
dinner_menu = {"스테이크", "리조또", "샐러드", "디저트"}
# 새로운 요리 추가
lunch_menu.add("샌드위치")
dinner_menu.add("피자")
# 특정 요리 제거
lunch_menu.remove("스프")
# 메뉴 교집합 (점심과 저녁 모두에 있는 요리)
common_dishes = lunch_menu.intersection(dinner_menu)
# 메뉴 합집합 (모든 요리)
all_dishes = lunch_menu.union(dinner_menu)
# 점심 메뉴에서 저녁 메뉴 제외한 요리
lunch_only = lunch_menu.difference(dinner_menu)
# 대칭 차집합 (한쪽 메뉴에만 있는 요리)
unique_dishes = lunch_menu.symmetric_difference(dinner_menu)
print("공통 요리:", common_dishes)
print("모든 요리:", all_dishes)
print("점심 전용 요리:", lunch_only)
print("특정 메뉴 전용 요리:", unique_dishes)
```

```
# 3.5 파이썬 실습
# 3.5.2 함수의 그래픽 표현
#
import matplotlib.pyplot as plt
import numpy as np

# 데이터 범위 설정 (년)
years = np.linspace(0, 10, 100)

# 선형 함수: 부동산 가격 y = 1000*x + 20000 (기초 가격 2만 달러, 연간 증가 1000 달러)
price_linear = 1000 * years + 20000

# 2차 함수: y = 3000*(x^2) - 1000*x + 20000 (변동성이 큰 시장)
price_quadratic = 3000 * (years ** 2) - 1000 * years + 20000

# 그래프 그리기
plt.figure(figsize=(12, 6))

# 선형 함수 그래프
plt.subplot(1, 2, 1)
plt.plot(years, price_linear, label="Linear")
plt.title("Linear Real Estate Price")
plt.xlabel("Years")
plt.ylabel("Price (USD)")
plt.legend()

# 2차 함수 그래프
plt.subplot(1, 2, 2)
plt.plot(years, price_quadratic, label="Quadratic")
plt.title("Quadratic Real Estate Price")
plt.xlabel("Years")
plt.ylabel("Price (USD)")
plt.legend()

# 전체 그래프 보여주기
plt.tight_layout()
plt.show()
```

```
# 3.5 파이썬 실습
# 3.5.2 함수의 그래픽 표현
# 다항함수의 그래프
#
import matplotlib.pyplot as plt
import numpy as np
# 데이터 범위 설정
x = np.linspace(-5, 10, 400)
# 5차 다항 함수
y_quintic = x**5 - 4*x**4 - 20*x**3 + 44*x**2 + 48*x - 48
# 그래프 그리기
plt.figure(figsize=(6, 4))
# 5차 다항 함수 그래프
plt.plot(x, y_quintic, label="Quintic")
plt.title("5th Degree Polynomial with Distinct Peaks and Valleys")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.ylim(-800, 400)  # y축 범위 설정
plt.text(10, 150, r'$y = \ x^5 - 4x^4 - 20x^3 + 44x^2 + 48x - 48$', fontsize=12, verticalalignment='bottom', horizontalalignment='right')
plt.legend()
# 전체 그래프 보여주기
plt.tight_layout()
plt.show()
```

```
# 3.5 파이썬 실습
# 3.5.3 수열과 급수 계산
# 등차수열과 등비수열
#
# 등차수열: 매년 5000달러를 저축
def arithmetic_sequence(a, d, n):
    return [a + d*i for i in range(n)]
# 등비수열: 매년 5%의 이자로 투자
def geometric_sequence(a, r, n):
    return [a * r**i for i in range(n)]
# 저축 예제: 시작값 5000달러, 매년 5000달러 추가, 10년간
savings_seq = arithmetic_sequence(5000, 5000, 10)
print("Savings Sequence:", savings_seq)
# 투자 예제: 시작값 5000달러, 연이율 5%, 10년간
investment_seq = geometric_sequence(5000, 1.05, 10)
print("Investment Sequence:", investment_seq)
```

```
# 3.5 파이썬 실습
# 3.5.3 수열과 급수 계산
# 급수
#
# 등차급수 계산 함수
def arithmetic_series(a, d, n):
    return n/2 * (2*a + (n - 1)*d)
# 등비급수 계산 함수
def geometric_series(a, r, n):
    if r == 1:
        return n * a
    else:
        return a * (1 - r**n) / (1 - r)
# 등차급수 예제: 시작값 1, 공차 3, 첫 5항의 합
arithmetic_sum = arithmetic_series(1, 3, 5)
print("Arithmetic Series Sum:", arithmetic_sum)
# 등비급수 예제: 시작값 2, 공비 3, 첫 4항의 합
geometric_sum = geometric_series(2, 3, 4)
print("Geometric Series Sum:", geometric_sum)
```

```
# 3.5 파이썬 실습
# 3.5.4 지수 및 로그함수 시각화
# 지수함수와 로그함수의 그래프
#
import matplotlib.pyplot as plt
import numpy as np
# 데이터 범위 설정 (일)
days_exp = np.linspace(0, 50, 400)
days_log = np.linspace(1, 50, 400)
# 지수함수: y = e^(0.1*x) (확산 모델)
infected_population = np.exp(0.1 * days_exp)
# 로그함수: y = log2(x) (방역 모델)
contained_population = np.log2(days_log)
# 그래프 그리기
plt.figure(figsize=(12, 6))
# 지수함수 그래프
plt.subplot(1, 2, 1)
plt.plot(days_exp, infected_population, label="Exponential")
plt.title("Infection Spread (Exponential Growth)")
plt.xlabel("Days")
plt.ylabel("Infected Population")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.ylim(-5, 300)  # y축 범위 설정
plt.text(30, 150, r'$y = e^{0.1x}$', fontsize=12, verticalalignment='bottom', horizontalalignment='right')
plt.legend()
# 로그함수 그래프
plt.subplot(1, 2, 2)
plt.plot(days_log, contained_population, label="Logarithmic")
plt.title("Containment Effort (Logarithmic)")
plt.xlabel("Days")
plt.ylabel("Contained Population")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.ylim(-1, 6)  # y축 범위 설정
plt.text(30, 4, r'$y = \log_2(x)$', fontsize=12, verticalalignment='bottom', horizontalalignment='right')
plt.legend()
# 전체 그래프 보여주기
plt.tight_layout()
plt.show()

```
