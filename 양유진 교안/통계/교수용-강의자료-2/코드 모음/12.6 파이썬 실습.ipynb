```python
# 12.6 파이썬 실습
# 12.6.1 모평균 추정
# 모평균에 대한 95% 신뢰도로 구간추정
#
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt
# 실제 50개의 표본 데이터
sample_data = np.random.normal(200, 20, 50)  # 예시로 평균 200, 표준편차 20인 정규분포에서 표본을 생성
# 표본 평균과 표본 표준편차 계산
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)  # ddof=1로 설정하여 표본 표준편차 계산
# 표본 크기
n = len(sample_data)
# z-값 계산 (95% 신뢰수준)
z_score = norm.ppf(0.975)  # (1 - 0.05/2) = 0.975
# 신뢰구간 계산
margin_of_error = z_score * (sample_std / np.sqrt(n))
lower_bound = sample_mean - margin_of_error
upper_bound = sample_mean + margin_of_error
# 결과 출력
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Sample Standard Deviation: {sample_std:.2f}")
print(f"95% Confidence Interval: ({lower_bound:.2f}, {upper_bound:.2f})")
print(f"Margin of Error: {margin_of_error:.2f}")
# 시각화
plt.figure(figsize=(10, 6))
plt.hist(sample_data, bins=20, edgecolor='k', alpha=0.7, density=True, label='Sample Data')
plt.axvline(lower_bound, color='red', linestyle='--', label=f'Lower Bound ({lower_bound:.2f})')
plt.axvline(upper_bound, color='red', linestyle='--', label=f'Upper Bound ({upper_bound:.2f})')
plt.axvline(sample_mean, color='green', linestyle='-', label=f'Sample Mean ({sample_mean:.2f})')
plt.title('95% Confidence Interval for the Mean Weight')
plt.xlabel('Weight')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 12.6 파이썬 실습
# 12.6.1 모평균 추정
# 모분산을 모르고 있을 때 모평균에 대한 99% 신뢰도로 구간추정
#
import numpy as np
from scipy.stats import t

# 주어진 데이터
n = 30  # 표본 크기
mean_score = 85  # 표본 평균
s = 15  # 표본 표준편차

# 자유도
df = n - 1

# t-값 계산 (99% 신뢰수준)
t_score = t.ppf(0.995, df)  # (1 - 0.01/2) = 0.995

# 신뢰구간 계산
margin_of_error = t_score * (s / np.sqrt(n))
lower_bound = mean_score - margin_of_error
upper_bound = mean_score + margin_of_error

# 결과 출력
print(f"99% Confidence Interval: ({lower_bound:.2f}, {upper_bound:.2f})")
print(f"Margin of Error: {margin_of_error:.2f}")
```

```python
# 12.6 파이썬 실습
# 12.6.2 모비율 추정
# 모비율의 95% 신뢰구간
#
import numpy as np
from scipy.stats import norm

# 주어진 데이터
n = 49  # 표본 크기
x = 30  # 긍정적인 반응을 보인 사람의 수
p_hat = x / n  # 표본비율

# 신뢰구간 계산을 위한 z-값 (95% 신뢰수준)
z_score = norm.ppf(0.975)  # (1 - 0.05/2) = 0.975

# 표본비율의 표준오차 계산
se = np.sqrt(p_hat * (1 - p_hat) / n)

# 신뢰구간 계산
margin_of_error = z_score * se
lower_bound = p_hat - margin_of_error
upper_bound = p_hat + margin_of_error

# 결과 출력
print(f"95% Confidence Interval for the proportion: ({lower_bound:.4f}, {upper_bound:.4f})")
print(f"Margin of Error: {margin_of_error:.4f}")
```

```python
# 12.6 파이썬 실습
# 12.6.2 모비율 추정
# 모비율의 90% 신뢰구간
#
import numpy as np
from scipy.stats import t

# 주어진 데이터
n = 24  # 표본 크기
x = 15  # 정책을 지지하는 사람의 수
p_hat = x / n  # 표본비율

# 자유도
df = n - 1

# t-값 계산 (90% 신뢰수준)
t_score = t.ppf(0.95, df)  # (1 - 0.10/2) = 0.95

# 표본비율의 표준오차 계산
se = np.sqrt(p_hat * (1 - p_hat) / n)

# 신뢰구간 계산
margin_of_error = t_score * se
lower_bound = p_hat - margin_of_error
upper_bound = p_hat + margin_of_error

# 결과 출력
print(f"90% Confidence Interval for the proportion: ({lower_bound:.4f}, {upper_bound:.4f})")
print(f"Margin of Error: {margin_of_error:.4f}")
```
