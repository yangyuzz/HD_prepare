```python
# 2.3 matplotlib 개요
# 2.3.1 matplotlib 가져오기(임포트)
#
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 2.3 matplotlib 개요
# 2.3.2 데이터를 선 그래프로 나타내기
# 제중 측정 데이터 예로 선 그래프 그리기
#
import numpy as np
import matplotlib.pyplot as plt
x = np.array([1, 2, 3, 4, 5, 6, 7]) # 날짜 데이터
y = np.array([78.4, 78.6, 78.5, 78.9, 78.2, 79.4, 78.6]) # 체중 데이터
plt.plot(x, y, color="r") # 빨간색 선으로 그래프를 나타낸다
plt.show() # 그래프를 보여준다
```

```python
# 2.3 matplotlib 개요
# 2.3.2 데이터를 선 그래프로 나타내기
# 체중 측정 데이터 예의 산포도 그리기
#
import numpy as np
import matplotlib.pyplot as plt
x = np.array([1, 2, 3, 4, 5, 6, 7])
y = np.array([78.4, 78.6, 78.5, 78.9, 78.2, 79.4, 78.6])
plt.scatter(x, y, color="r") # 빨간색으로 산포도 형식으로 보여준다
plt.show()
```

```python
# 2.3 matplotlib 개요
# 2.3.2 데이터를 선 그래프로 나타내기
# 체중 측정 데이터 예에 대한 grid()함수 적용
#
import numpy as np
import matplotlib.pyplot as plt
x = np.array([1, 2, 3, 4, 5, 6, 7])
y = np.array([78.4, 78.6, 78.5, 78.9, 78.2, 79.4, 78.6])
plt.plot(x, y, color='r',marker='o') # 색은 빨간색, 점을 추가해서 그래프로 나타낸다
plt.grid() # 그리드(격자) 형식으로 나타낸다
plt.show()
```

```python
# 2.3 matplotlib 개요
# 2.3.2 데이터를 선 그래프로 나타내기
# NumPy의 linspace() 함수 적용
#
import numpy as np

x = np.linspace(-3, 3) # -3에서 3까지 50개의 구간으로 나눈다
print(x)
print()
print(len(x)) # x의 요소 수
```

```python
# 2.3 matplotlib 개요
# 2.3.2 데이터를 선 그래프로 나타내기
# NumPy의 np.linspace(시작 값, 마지막 값, 샘플 개수) 형태를 실제 그래프에 적용
#
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace (-5, 5, 300) # -5에서 5까지 300개의 구간으로 나눈다
yl = x ** 2
y2 = (x - 2) ** 2
plt.plot(x, yl, color = "r") # 선은 빨간색
plt.plot(x, y2, color = "k", linestyle = "--") #색은 검정색, 선 스타일은 점선
plt.grid()
plt.show()
```

```python
# 2.3 matplotlib 개요
# 2.3.3 그래프 모양 꾸미기
# 그래프를 이해하기 쉽도록 하는 디자인
#
# 1. 나눔 글꼴 설치 (Colab에는 기본 한글 폰트가 없으므로 수동 설치 필요)
!apt-get update -qq
!apt-get install -y fonts-nanum

# 2. 필요한 라이브러리 불러오기
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import numpy as np

# 3. 나눔 글꼴 경로 설정 및 matplotlib에 등록
font_path = "/usr/share/fonts/truetype/nanum/NanumGothic.ttf"
fm.fontManager.addfont(font_path)

# 4. matplotlib에 한글 폰트 적용
plt.rc('font', family=fm.FontProperties(fname=font_path).get_name())

# 5. 마이너스 기호가 깨지지 않도록 설정
plt.rcParams["axes.unicode_minus"] = False

# 6. 데이터 생성
x = np.linspace(-5, 5)
y1 = 2 * x
y2 = 5 * x

# 7. 그래프 그리기
plt.plot(x, y1, label="y1 값", color='r')  # 빨간 선
plt.plot(x, y2, label="y2 값", linestyle="dashed", color='k')  # 검정 점선

# 8. 축, 제목, 범례, 그리드 설정
plt.xlabel("X-축", size=14)
plt.ylabel("Y-축", size=14)
plt.title("그래프 제목")
plt.grid()
plt.legend()

# 9. 그래프 출력
plt.show()
```

```python
# 2.3 matplotlib 개요
# 2.3.4 히스토그램 그리기
# 히스토그램 그리기
#
import numpy as np
import matplotlib.pyplot as plt

# 데이터 정의
data = np.array([60, 71, 62, 62, 62, 62, 71, 64, 68, 64, 66, 66, 67, 67, 68, 68, 68, 69, 60, 71])

# 히스토그램 그리기
plt.hist(data, bins=20, color='skyblue', edgecolor='black', label='점수 분포')

# 축 레이블 설정
plt.xlabel("점수", fontsize=12)
plt.ylabel("빈도", fontsize=12)

# 제목 설정
plt.title("시험 점수 분포 히스토그램", fontsize=14)

# 그리드 추가
plt.grid(True, linestyle='--', alpha=0.6)

# 범례 추가
plt.legend()

# 그래프 출력
plt.show()
```

```python
# 2.3 matplotlib 개요
# 2.3.4 히스토그램 그리기
# 다양한 히스토그램 그리기
#
import numpy as np
import matplotlib.pyplot as plt

weight1 = np.array([60, 71, 62, 62, 62, 62, 71, 65, 68, 65, 66, 66, 67, 67, 68, 68, 68, 69, 60, 71])
weight2 = np.array([63, 71, 63, 63, 65, 65, 68, 63, 65, 67, 69, 62, 68, 62, 69, 62, 65, 67, 64, 63])

fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(8, 8))

axes[0, 0].hist((weight1, weight2), histtype='bar')
axes[0, 0].set_title('histtype - bar')

axes[0, 1].hist((weight1, weight2), histtype='barstacked')
axes[0, 1].set_title('histtype - barstacked')

axes[1, 0].hist((weight1, weight2), histtype='step')
axes[1, 0].set_title('histtype - step')

axes[1, 1].hist((weight1, weight2), histtype='stepfilled')
axes[1, 1].set_title('histtype - stepfilled')
plt.tight_layout()
plt.show()
```
