```python
# 8.6 파이썬 실습
# 8.6.1 파이썬을 사용한 확률변수의 분포 시각화
# 확률질량함수와 그래프 그리기
#
import matplotlib.pyplot as plt
from collections import Counter
# 두 개의 공정한 주사위 던지기 시뮬레이션
dice_rolls = [(i, j) for i in range(1, 7) for j in range(1, 7)]
sums = [sum(roll) for roll in dice_rolls]
# 확률변수 X의 확률질량함수 계산
f_X = Counter(sums)
total_outcomes = len(dice_rolls)
f_X = {x: count/total_outcomes for x, count in f_X.items()}
# 확률질량함수 값을 점으로 표현하는 그래프로 그리기
plt.figure(figsize=(8, 6))
plt.plot(list(f_X.keys()), list(f_X.values()), 'bo')  # 점
plt.vlines(list(f_X.keys()), 0, list(f_X.values()), colors='blue')  # 수선의 발
plt.title('Probability Mass Function of X')
plt.xlabel('Sum of dice rolls (X)')
plt.ylabel('Probability')
plt.xticks(range(2, 13))  # X의 가능한 값은 2부터 12까지
plt.ylim(0, 1)  # 확률의 범위는 0부터 1
plt.grid(axis='y')
plt.show()
```

```python
# 8.6 파이썬 실습
# 8.6.2 기댓값과 분산의 계산
# 기댓값과 분산
#
# 확률변수 Y의 점수와 그 점수를 얻을 확률
scores = [0, 1, 2, 3, 4]
probabilities_Y = [0.1, 0.2, 0.4, 0.2, 0.1]

# 기댓값 계산
expected_value_Y = sum(y * p for y, p in zip(scores, probabilities_Y))

# 분산 계산
variance_Y = sum((y - expected_value_Y) ** 2 * p for y, p in zip(scores, probabilities_Y))

expected_value_Y, variance_Y
```

```python
# 8.6 파이썬 실습
# 8.6.3 결합분포 및 상관관계 분석
# 주변확률질량함수와 그래프
#
import matplotlib.pyplot as plt

# 결합확률질량함수 데이터
joint_pmf_data = {
    (1, 1): 0.1,
    (1, 2): 0.2,
    (1, 3): 0.2,
    (2, 1): 0.3,
    (2, 2): 0.05,
    (2, 3): 0.15
}

# X와 Y의 주변확률밀도함수를 계산
f_X = {}
f_Y = {}

# X의 주변확률밀도함수 계산
for x, y in joint_pmf_data.keys():
    f_X[x] = f_X.get(x, 0) + joint_pmf_data[(x, y)]

# Y의 주변확률밀도함수 계산
for x, y in joint_pmf_data.keys():
    f_Y[y] = f_Y.get(y, 0) + joint_pmf_data[(x, y)]

# Y축의 범위를 0부터 1까지, X축의 범위를 동일하게 설정하여 그래프를 그리기
plt.figure(figsize=(12, 5))

# Y축의 범위 설정
y_min = 0
y_max = 1

# X축의 범위 설정
x_min = 0.5
x_max = 3.5

# X의 주변확률밀도함수 그래프
plt.subplot(1, 2, 1)
plt.plot(list(f_X.keys()), list(f_X.values()), 'bo')  # 점
plt.vlines(list(f_X.keys()), y_min, list(f_X.values()), colors='blue')  # 수선의 발
plt.title('Marginal PMF of X')
plt.xlabel('X value')
plt.ylabel('Probability')
plt.xticks(range(int(x_min), int(x_max)+1))
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)

# Y의 주변확률밀도함수 그래프
plt.subplot(1, 2, 2)
plt.plot(list(f_Y.keys()), list(f_Y.values()), 'ro')  # 점
plt.vlines(list(f_Y.keys()), y_min, list(f_Y.values()), colors='red')  # 수선의 발
plt.title('Marginal PMF of Y')
plt.xlabel('Y value')
plt.ylabel('Probability')
plt.xticks(range(int(x_min), int(x_max)+1))
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)

plt.tight_layout()
plt.show()
```

```python
# 8.6 파이썬 실습
# 8.6.3 결합분포 및 상관관계 분석
# 공분산
#
import numpy as np

# 국어 성적과 수학 성적 데이터
korean_scores = [75, 65, 50, 70, 80]
math_scores = [65, 70, 70, 55, 60]

# 데이터를 NumPy 배열로 변환
data = np.array([korean_scores, math_scores])

# 공분산 계산
covariance = np.cov(data, bias=True)[0, 1]

# 결과 출력
print("Covariance:", covariance)
```

```python
# 8.6 파이썬 실습
# 8.6.3 결합분포 및 상관관계 분석
# 상관계수
#
import numpy as np

import numpy as np

# 데이터
x = np.array([4, 6, -4, 10])
y = np.array([-1, 5, 3, 18])

# 상관계수 계산
correlation = np.corrcoef(x, y)[0, 1]

# 결과 출력
print("Correlation Coefficient:", correlation)
```
