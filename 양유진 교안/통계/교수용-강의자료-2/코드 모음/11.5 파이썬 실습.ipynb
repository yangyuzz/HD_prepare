```python
# 11.5 파이썬 실습
# 11.5.1 표본평균의 분포 시뮬레이션
# 표본평균의 히스토그램
#
import numpy as np
import matplotlib.pyplot as plt

# 모집단 설정
population_mean = 50
population_std = 10

# 표본추출 설정
sample_size = 30
number_of_samples = 1000

# 표본평균 저장을 위한 배열
sample_means = []

# 표본추출 및 표본평균 계산
for _ in range(number_of_samples):
    sample = np.random.normal(population_mean, population_std, sample_size)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)

# 표본평균의 분포 그리기
plt.hist(sample_means, bins=30, color='blue', edgecolor='black', alpha=0.7)
plt.title('Distribution of Sample Means')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()

# 결과 출력
print(f"Mean of sample means: {np.mean(sample_means):.2f}")
print(f"Standard deviation of sample means: {np.std(sample_means, ddof=1):.2f}")
```

```python
# 11.5 파이썬 실습
# 11.5.2 중심극한정리의 실제 데이터 적용
# 중심극한정리
#
import numpy as np
import matplotlib.pyplot as plt
# 균일 분포의 매개변수 설정
a = 0
b = 1
# 표본 크기 및 반복 횟수 설정
sample_sizes = [1, 5, 30, 100]
repetitions = 1000
# 그래프 그리기
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
axes = axes.ravel()
for i, sample_size in enumerate(sample_sizes):
    # 표본평균 저장을 위한 배열
    sample_means = []

    # 데이터 생성 및 표본평균 계산
    for _ in range(repetitions):
        sample = np.random.uniform(a, b, sample_size)
        sample_mean = np.mean(sample)
        sample_means.append(sample_mean)

    # 표본평균의 분포 그리기
    axes[i].hist(sample_means, bins=30, color='blue', alpha=0.7, density=True)
    axes[i].set_title(f'Sample Size = {sample_size}')
    axes[i].set_xlabel('Sample Mean')
    axes[i].set_ylabel('Density')
    axes[i].grid(True)
plt.tight_layout()
plt.show()
```

```python
# 11.5 파이썬 실습
# 11.5.3 표본비율 분포의 계산 및 시각화
# 표본비율 분포 시각화
#
import numpy as np
import matplotlib.pyplot as plt

# 성공 확률 (예를 들어 '예'라고 답변한 비율)
p = 0.6

# 전체 응답자 수
population_size = 1000

# 표본 크기
sample_size = 100

# 시뮬레이션 반복 횟수
repetitions = 1000

# 표본비율 저장을 위한 배열
sample_proportions = []

# 데이터 생성 및 표본비율 계산
for _ in range(repetitions):
    sample = np.random.binomial(1, p, sample_size)
    sample_proportion = np.mean(sample)
    sample_proportions.append(sample_proportion)

# 표본비율의 분포 그리기
plt.hist(sample_proportions, bins=30, color='blue', edgecolor='black', alpha=0.7)
plt.title('Distribution of Sample Proportions')
plt.xlabel('Sample Proportion')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()

# 평균과 표준편차 출력
print(f"Mean of sample proportions: {np.mean(sample_proportions):.4f}")
print(f"Standard deviation of sample proportions: {np.std(sample_proportions, ddof=1):.4f}")
```
