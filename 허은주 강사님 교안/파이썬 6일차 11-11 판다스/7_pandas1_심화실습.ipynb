### 심화 문제 05. 배송 소요일과 날짜 오류 확인

주문일·배송일의 문자열을 날짜로 바꾸고 배송 소요일을 계산한다.

1. 두 날짜 열을 `pd.to_datetime(errors="coerce")`로 변환한다. 잘못된 날짜는 `NaT`가 된다.

3. 배송 소요일이 확인되는 행의 주문번호와 배송 소요일을 각각 리스트로 출력한다.

**예상 출력**

```text

배송소요일: [2.0, 3.0, 4.0]
계산 불가: 1
평균 소요일: 3.0
```

### 심화 문제 06. 주문 데이터 정리와 결과 저장

실제 분석 전에 주문 데이터의 코드·금액·날짜를 정리하고 배송 방식을 판정하여 CSV 보고서를 만든다.

1. `상품코드`에서 공백을 제거하고 영문을 대문자로 바꾼다.
2. `금액`을 `pd.to_numeric(errors="coerce")`로 변환하고, 변환할 수 없는 금액은 0으로 채운다.
3. `주문일`을 날짜로 바꾸고 `.dt.month`로 `월` 열을 만든다.
4. `제품군`과 `금액` 열을 각각 리스트로 가져와 `for`문과 `if`문으로 배송 방식을 결정한 뒤 `배송` 열에 저장한다.
   - 금액이 **100000 이상**이면 `"직배송"`으로 한다.
   - 그렇지 않고 제품군이 `"가구"`이면 `"화물"`로 한다.
   - 그 밖의 경우는 `"택배"`로 한다.
5. `상품코드`, `금액`, `월`, `배송`을 각각 리스트로 출력한다.
6. `orders`를 `심화06_주문정리.csv`로 저장한 다음 다시 읽어 행 수와 금액 합계를 출력한다. (`index=False`, `encoding="utf-8-sig"`)

**예상 출력**

```text
상품코드: ['AA-01', 'BB-02', 'CC-03']
금액: [85000.0, 150000.0, 0.0]
월: [6, 7, 7]
배송: ['택배', '직배송', '택배']
불러온 행 수: 3
금액 합계: 235000.0
```
