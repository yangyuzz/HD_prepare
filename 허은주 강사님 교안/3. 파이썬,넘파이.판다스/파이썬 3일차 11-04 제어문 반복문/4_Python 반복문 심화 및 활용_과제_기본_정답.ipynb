# 정답 확인

먼저 스스로 풀고 비교한다.

### 1. range() 정답

````python
# 문제 1-1
for i in range(4):              # 0, 1, 2, 3
    print(f"{i + 1}번 라인 가동 점검")

total = 0
for number in range(1, 11):     # 1부터 10까지 (11은 포함하지 않는다)
    total += number
print(f"1부터 10까지의 합: {total}")

for hour in range(9, 18, 3):    # 9부터 3씩 증가, 18 전까지
    print(f"{hour}시 온도 측정")
````

### 2. 컴프리헨션 정답

````python
# 문제 2-1
prices_text = ["15000", "8200", "23500", "4900", "12000"]

prices = [int(p) for p in prices_text]
print(prices)

over_10000 = [p for p in prices if p >= 10000]   # 뒤의 if 로 조건에 맞는 값만 담는다
print(over_10000)

with_vat = [round(p * 1.1) for p in prices]
print(with_vat)
````

````python
# 문제 2-2
parts = [" bt-10", "BT-10", "nt-07 ", "Nt-07", "ws-22", " BT-10 "]

unique_parts = {p.strip().upper() for p in parts}   # 중괄호를 쓰면 집합이 되어 중복이 사라진다

print(sorted(unique_parts))
print(f"부품 종류: {len(unique_parts)}개")
````

### 3. 딕셔너리 반복 정답

````python
# 문제 3-1
region_sales = {"서울": 350000, "부산": 280000, "대구": 210000}

for region, value in region_sales.items():   # (키, 값) 튜플을 바로 언패킹한다
    print(f"{region}: {value:,}원")

for region, value in region_sales.items():
    if value < 300000:
        print(f"{region}: 목표 미달")
````

### 4. get(키, 0)으로 개수·합계 세기 정답

````python
# 문제 4-1
defect_types = ["스크래치", "찍힘", "스크래치", "변색", "찍힘", "스크래치", "이물"]

counts = {}
for defect in defect_types:
    counts[defect] = counts.get(defect, 0) + 1   # 처음 나온 유형은 0에서 시작

for defect, count in counts.items():
    print(f"{defect}: {count}건")
````

````python
# 문제 4-2
shipments = [("울산", 120), ("창원", 85), ("울산", 64), ("구미", 150), ("창원", 40)]

totals = {}
for city, quantity in shipments:
    totals[city] = totals.get(city, 0) + quantity

for city, total in totals.items():
    print(f"{city}: {total}개")
````

### 5. enumerate()와 zip() 정답

````python
# 문제 5-1
weekly = [980, 1120, 860, 1045, 1010]

for day, quantity in enumerate(weekly, start=1):   # start 를 생략하면 0부터
    print(f"{day}일차: {quantity:,}개")

for day, quantity in enumerate(weekly, start=1):
    if quantity < 1000:
        print(f"{day}일차 목표 미달 ({quantity}개)")
````

````python
# 문제 5-2
regions = ["서울", "부산", "대전"]
sales = [350000, 280000, 210000]

for region, value in zip(regions, sales):   # 같은 위치의 값끼리 묶인다
    print(f"{region}: {value:,}원")

sales_dict = dict(zip(regions, sales))
print(sales_dict)
````
