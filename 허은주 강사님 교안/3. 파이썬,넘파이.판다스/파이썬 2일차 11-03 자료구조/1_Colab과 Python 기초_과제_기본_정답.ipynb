# 정답 확인

### 1. 변수와 출력 정답

````python
# 문제 1-1
branch = "강남점"
quantity = 48
unit_price = 15000
total_sales = quantity * unit_price

print(f"{branch}의 총매출은 {total_sales:,}원이다.")
````

````python
# 문제 1-2
line_name = "A라인"
produced = 18450
defects = 412
work_hours = 8

per_hour = produced / work_hours
defect_rate = defects / produced

print(f"{line_name} 생산량: {produced:,}개")
print(f"시간당 생산량: {per_hour:,.2f}개")
print(f"불량률: {defect_rate:.2%}")   # % 서식은 100을 곱해 백분율로 보여 준다
````

### 2. 자료형과 형변환 정답

````python
# 문제 2-1
price_text = "23000"
quantity_text = "7"

price = int(price_text)
quantity = int(quantity_text)

print(type(price_text), type(price))   # 변환 전은 str, 변환 후는 int

total = price * quantity
print(f"총금액: {total:,}원")
````

### 3. 산술 연산 정답

````python
# 문제 3-1
produced = 250
box_size = 18
pack_cost = 1200

import math

full_boxes, remainder = divmod(produced, box_size)
needed_boxes = math.ceil(produced / box_size)   # 13.88... 을 올려 14
total_cost = needed_boxes * pack_cost

print(f"가득 찬 상자: {full_boxes}개, 남는 수량: {remainder}개")
print(f"필요한 상자: {needed_boxes}개")
print(f"포장비: {total_cost:,}원")
````

### 4. 비교·논리 연산자와 in 정답

````python
# 문제 4-1
sales = 250000
region = "서울"
defect_codes = ["D01", "D04", "D07"]

print(sales >= 200000 and region == "서울")   # 두 조건 모두 참
print(sales >= 300000 or region == "부산")    # 두 조건 모두 거짓
print("D04" in defect_codes)
print("D02" not in defect_codes)
````

````python
# 문제 5-3
lot_no = "LT20260415-B07"

# 인덱스:  L(0) T(1) 2(2) 0(3) 2(4) 6(5) 0(6) 4(7) 1(8) 5(9) -(10) B(11) 0(12) 7(13)
print(lot_no[:2])                          # 처음부터 1번까지
print(lot_no[2:6])                         # 2번부터 5번까지 (끝 번호 6은 포함하지 않는다)
print(f"{lot_no[6:8]}월 {lot_no[8:10]}일")
print(lot_no[-1])                          # 뒤에서 첫 번째
print(lot_no[-3:])                         # 뒤에서 세 번째부터 끝까지
````

### 6. 리스트 정답

````python
# 문제 6-1
branches = ["강남점", "서초점", "송파점", "분당점"]

branches.append("마포점")    # 맨 뒤에 추가
branches.remove("서초점")    # 값으로 찾아 삭제

print(branches)
print("송파점" in branches)
````

````python
# 문제 6-2
sales = [92000, 145000, 88000, 220000, 167000, 133000, 205000]

print(f"첫 매출: {sales[0]:,}원")
print(f"마지막 매출: {sales[-1]:,}원")
print(f"2~4번째 매출: {sales[1:4]}")
print(f"마지막 세 건: {sales[-3:]}")

sorted_sales = sorted(sales, reverse=True)
print(sorted_sales)
print(sales)
print(f"최고 매출: {sorted_sales[0]:,}원")
print(f"최저 매출: {sorted_sales[-1]:,}원")
````
