### 기본 문제 19. nlargest()와 nsmallest() — 정답

````python
import pandas as pd
vendors = pd.DataFrame({"거래처":["A", "B", "C", "D", "E", "F"], "납품금액":[420000, 980000, 150000, 760000, 530000, 290000]})

print("상위 3:", vendors.nlargest(3, "납품금액")["거래처"].tolist())
print("하위 2:", vendors.nsmallest(2, "납품금액")["거래처"].tolist())
print("최대 금액:", vendors["납품금액"].max())
````

### 기본 문제 20. axis로 행별·열별 계산하기 — 정답

````python
import pandas as pd
output = pd.DataFrame({"1월":[200, 180, 210], "2월":[190, 175, 220], "3월":[205, 185, 215]}, index=["L1", "L2", "L3"])

print("월별 합계:")
print(output.sum(axis=0))
output["합계"] = output.sum(axis=1)
print("라인별 합계:", output["합계"].tolist())
print("1월 제외 열:", output.drop("1월", axis=1).columns.tolist())
````
