### 문제 1-4. 과목 점수 문자열로 성적표 만들기 — 정답

````python
report = "국어=82;영어=91;수학=75"

grades = {}
for item in report.split(";"):
    subject, score = item.split("=")
    grades[subject] = int(score)

highest = max(grades.values())
for subject, score in grades.items():
    if score == highest:
        best_subject = subject
        break
print(f"과목 수: {len(grades)}")
print(f"총점: {sum(grades.values())}점")
print(f"평균: {sum(grades.values()) / len(grades):.1f}점")
print(f"최고점 과목: {best_subject} ({highest}점)")
````

## 2. 리스트·튜플·집합 활용 정답

### 문제 2-1. 매출과 환불 내역 분리하기 — 정답

````python
amounts = [12000, -5000, 33000, 0, -2000, 14000]

sales = []
refunds = []
for amount in amounts:
    if amount > 0:
        sales.append(amount)
    elif amount < 0:
        refunds.append(abs(amount))
print(f"매출 내역: {sales}")
print(f"환불 내역: {refunds}")
print(f"매출 합계: {sum(sales):,}원")
print(f"환불 합계: {sum(refunds):,}원")
````

## 5. 함수·내장함수·예외 처리 정답

### 문제 5-1. 근무시간에 따라 기본급·연장급 계산 — 정답

````python
def work_pay(hours, hourly=12000):
    normal = min(hours, 8)
    overtime = max(hours - 8, 0)
    return int(normal * hourly + overtime * hourly * 1.5)

print(f"6시간 급여: {work_pay(6):,}원")
print(f"10시간 급여: {work_pay(10):,}원")
````

### 문제 5-2. 점수 요약 함수를 만들어 빈 리스트도 처리 — 정답

````python
scores = [95, 72, 88, 85]

def summarize(values):
    if not values:
        return None
    return {"인원": len(values), "평균": sum(values) / len(values),
            "최고": max(values), "최저": min(values)}

result = summarize(scores)
print(f"인원: {result['인원']}명")
print(f"평균: {result['평균']:.1f}점")
print(f"최고/최저: {result['최고']}점 / {result['최저']}점")
print(f"빈 값: {summarize([])}")
````

### 문제 5-3. 상품코드 정리 함수 만들기 — 정답

````python
codes = [" a01 ", "A01", " b02 ", "", "B02", " c03"]

def normalize_codes(values):
    result = []
    for value in values:
        code = value.strip().upper()
        if code and code not in result:
            result.append(code)
    return result

cleaned = normalize_codes(codes)
print(f"정리한 코드: {cleaned}")
print(f"코드 개수: {len(cleaned)}개")
````

### 문제 5-4. 문자열 두 개로 안전한 나눗셈 — 정답

````python
pairs = [("120", "4"), ("50", "0"), ("abc", "2")]

def safe_divide(a_text, b_text):
    try:
        a = int(a_text)
        b = int(b_text)
        return f"{a / b:.2f}"
    except ValueError:
        return "숫자가 아니다"
    except ZeroDivisionError:
        return "0으로 나눌 수 없다"

for left, right in pairs:
    print(f"{left} / {right} = {safe_divide(left, right)}")
````

### 문제 5-5. 쉼표로 입력한 점수를 함수로 분석 — 정답

````python
def analyze_scores(text):
    scores = []
    for item in text.split(","):
        scores.append(int(item.strip()))
    passed = 0
    for score in scores:
        if score >= 80:
            passed += 1
    return len(scores), sum(scores) / len(scores), passed

text = input("점수(쉼표 구분): ")
count, average, passed = analyze_scores(text)
print(f"점수 개수: {count}개")
print(f"평균: {average:.2f}점")
print(f"80점 이상: {passed}명")
````
