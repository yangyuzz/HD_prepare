먼저 문제를 직접 풀고 정답과 비교한다. 아래 코드는 각 문제를 독립 실행할 수 있도록 준비 데이터를 포함한다.

### 심화 문제 01. 라인별 생산 품질 보고서 — 정답

````python
import pandas as pd
production = pd.Series([200, 150, 180], index=["A", "B", "C"], name="생산량")
defects = pd.Series([4, 9, 3], index=["A", "B", "C"], name="불량수")

report = pd.concat([production, defects], axis=1)
report["양품수"] = report["생산량"] - report["불량수"]
report["불량률(%)"] = (report["불량수"] / report["생산량"] * 100).round(1)
high = report[report["불량률(%)"] > 3]
print("관리 대상:", list(high.index))
print("불량률:", high["불량률(%)"].tolist())
print("전체 생산량:", int(report["생산량"].sum()))
print("전체 양품수:", int(report["양품수"].sum()))
````

### 심화 문제 02. 검사 기록 CSV 왕복 검증 — 정답

````python
import pandas as pd
checks = pd.DataFrame({"작업번호":["W1", "W2", "W1", "W3", "W4"], "검사수":[100.0, 80.0, 120.0, None, 90.0]})

checks.to_csv("심화02_검사.csv", index=False, encoding="utf-8-sig")
loaded = pd.read_csv("심화02_검사.csv")
print("크기 동일:", checks.shape == loaded.shape)
clean = loaded.drop_duplicates("작업번호", keep="last").copy()
clean["검사수"] = clean["검사수"].fillna(0)
print("작업번호:", clean["작업번호"].tolist())
print("검사수:", clean["검사수"].tolist())
````

### 심화 문제 03. 최근 점검 기록의 결측 처리 — 정답

````python
import pandas as pd
logs = pd.DataFrame({"설비":["A", "B", "A", "C", "B", "D"], "점수":[65.0, 81.0, 75.0, None, 83.0, 91.0], "판정":["주의", "정상", "정상", None, "정상", "정상"]})

latest = logs.drop_duplicates("설비", keep="last").copy()
latest["점수"] = latest["점수"].fillna(latest["점수"].median())
latest["판정"] = latest["판정"].fillna("확인필요")
print("설비:", latest["설비"].tolist())
print("점수:", latest["점수"].tolist())
print("판정:", latest["판정"].tolist())
print("남은 결측:", int(latest.isna().sum().sum()))
````

### 심화 문제 04. 오염된 제품 코드와 수량 정리 — 정답

````python
import pandas as pd
items = pd.DataFrame({"제품코드":[" mt-01 ", "ca-02 ", " mt-03", "pc-04"], "수량":["10", "오류", "25", "5"]})

items["제품코드"] = items["제품코드"].str.strip().str.upper()
items["수량"] = pd.to_numeric(items["수량"], errors="coerce")
valid = items.dropna(subset=["수량"]).copy()
print("제품코드:", valid["제품코드"].tolist())
print("수량:", valid["수량"].tolist())
print("수량 합계:", valid["수량"].sum())
````

### 심화 문제 05. 배송 소요일과 날짜 오류 확인 — 정답

````python
import pandas as pd
orders = pd.DataFrame({"주문번호":[101, 102, 103, 104], "주문일":["2026-05-01", "2026-05-03", "날짜오류", "2026-05-06"], "배송일":["2026-05-03", "2026-05-06", "2026-05-10", "2026-05-10"]})

orders["주문일"] = pd.to_datetime(orders["주문일"], errors="coerce")
orders["배송일"] = pd.to_datetime(orders["배송일"], errors="coerce")
orders["배송소요일"] = (orders["배송일"] - orders["주문일"]).dt.days
valid = orders.dropna(subset=["배송소요일"])
print("주문번호:", valid["주문번호"].tolist())
print("배송소요일:", valid["배송소요일"].tolist())
print("계산 불가:", int(orders["배송소요일"].isna().sum()))
print(f"평균 소요일: {valid['배송소요일'].mean():.1f}")
````

### 심화 문제 06. 주문 데이터 정리와 결과 저장 — 정답


````python
import pandas as pd
orders = pd.DataFrame({"상품코드":[" aa-01 ", "bb-02", " cc-03 "], "제품군":["전자", "가구", "전자"], "금액":["85000", "150000", "입력오류"], "주문일":["2026-06-01", "2026-07-04", "2026-07-10"]})

orders["상품코드"] = orders["상품코드"].str.strip().str.upper()
orders["금액"] = pd.to_numeric(orders["금액"], errors="coerce").fillna(0)
orders["주문일"] = pd.to_datetime(orders["주문일"])
orders["월"] = orders["주문일"].dt.month

print("상품코드:", orders["상품코드"].tolist())
print("금액:", orders["금액"].tolist())
print("월:", orders["월"].tolist())
print(f"100000 이상 주문: {(orders['금액'] >= 100000).sum()}건")
print(f"7월 주문: {(orders['월'] == 7).sum()}건")

orders.to_csv("심화06_주문정리.csv", index=False, encoding="utf-8-sig")
loaded = pd.read_csv("심화06_주문정리.csv")
print("불러온 행 수:", len(loaded))
print("금액 합계:", loaded["금액"].sum())
````
