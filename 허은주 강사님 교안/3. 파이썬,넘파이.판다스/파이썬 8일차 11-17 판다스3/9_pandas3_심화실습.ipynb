4. 그다음 `groupby()`로 카테고리별 주문금액 합계를 계산한다. `미등록` 카테고리도 결과에 포함한다.

### 심화 문제 05. agg·merge로 지역별 성과 비교

각 판매 기록에 지역 평균과 비교한 초과 금액을 붙이고, 지역별 이익률을 요약한다.

1. `groupby("지역", as_index=False).agg()`로 `매출합계`, `이익합계`, `건수`를 계산한다. 새 열 이름은 `agg()` 안에 `새열이름=("열", "계산")` 형태로 적는다. 예: `agg(매출합계=("매출", "sum"))`
2. 집계 표에 `이익률(%) = 이익합계/매출합계*100`을 소수 첫째 자리로 계산한다.
3. `groupby("지역", as_index=False)["매출"].mean()`으로 지역별 평균 표를 만든다. 원본과 `merge(on="지역", how="left")`로 합쳐 각 행의 평균을 연결한 뒤 `평균초과액=매출-지역평균매출`을 계산한다.
4. 지역별 이름·매출합계·이익률 리스트와 행별 평균초과액 리스트를 출력한다.

**힌트:** 그룹별 평균 표의 `매출` 열 이름을 `지역평균매출`로 바꾸면 병합 후 계산하기 쉽다.

**예상 출력**

```text
집계 지역: ['대전', '부산', '서울']
지역 매출: [150, 600, 400]
지역 이익률: [10.0, 18.3, 17.5]
행별 초과액: [-100.0, -100.0, 100.0, 100.0, 0.0]
```

````python
import pandas as pd
df = pd.DataFrame({"지역": ["서울", "부산", "서울", "부산", "대전"], "매출": [100, 200, 300, 400, 150], "이익": [10, 30, 60, 80, 15]})

# 여기에 작성한다.
````
