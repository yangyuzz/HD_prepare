# 정답 확인

먼저 스스로 풀고 비교한다.

### 문제 1. 시스템 기록 한 줄 정리하기 정답

````python
# 문제 1
record = "  2026-04-15|line-b|  3,840  |  96  "

parts = record.strip().split("|")
print(parts)   # 항목 안쪽의 공백은 그대로 남아 있다

date = parts[0].replace("-", "/")
line = parts[1].upper()
produced = int(parts[2].strip().replace(",", ""))   # "3,840" 은 쉼표 때문에 바로 int() 할 수 없다
defects = int(parts[3].strip())
defect_rate = defects / produced

print(f"날짜: {date}")
print(f"라인: {line}")
print(f"생산량: {produced:,}개 / 불량: {defects}개")
print(f"불량률: {defect_rate:.2%}")
````

### 문제 2. 숫자 문자열을 리스트로 바꿔 요약하기 정답

````python
# 문제 2
raw = " 1,250 ; 980 ;1,430; 1,108 ;1,320 "

clean = raw.replace(" ", "").replace(",", "")
parts = clean.split(";")
print(parts)   # 아직 문자열 리스트다

daily = [int(parts[0]), int(parts[1]), int(parts[2]), int(parts[3]), int(parts[4])]
print(daily)

total = daily[0] + daily[1] + daily[2] + daily[3] + daily[4]
average = total / len(daily)
print(f"합계: {total:,}개")
print(f"평균: {average:,.1f}개")

sorted_desc = sorted(daily, reverse=True)
print(f"상위 3일: {sorted_desc[:3]}")
print(f"최저 생산량: {sorted_desc[-1]:,}개")
````

### 문제 3. 출하 계획 계산하기 정답

````python
# 문제 3
order_a = 145
order_b = 230
order_c = 98
pallet_size = 40      # 팔레트 1개당 제품 수
truck_capacity = 5    # 트럭 1대당 팔레트 수

import math

total = 0
total += order_a
total += order_b
total += order_c
print(f"총 주문 수량: {total}개")

full_pallets, remainder = divmod(total, pallet_size)
print(f"가득 찬 팔레트: {full_pallets}개, 남는 제품: {remainder}개")

pallets = math.ceil(total / pallet_size)
print(f"필요한 팔레트: {pallets}개")

trucks = math.ceil(pallets / truck_capacity)
last_truck = pallets % truck_capacity
print(f"필요한 트럭: {trucks}대 (마지막 트럭 팔레트 {last_truck}개)")

max_load = trucks * truck_capacity * pallet_size   # 트럭 3대 × 팔레트 5개 × 제품 40개
load_rate = total / max_load
print(f"적재율: {load_rate:.1%}")
````

### 문제 5. 종합 문제: 5일 매출 요약 정답

````python
# 문제 5
daily_sales = [320000, 285000, 410000, 198000, 365000]

total = daily_sales[0] + daily_sales[1] + daily_sales[2] + daily_sales[3] + daily_sales[4]
average = round(total / len(daily_sales))   # 자릿수를 생략하면 정수로 반올림

sorted_sales = sorted(daily_sales, reverse=True)
best = sorted_sales[0]
worst = sorted_sales[-1]

print(f"5일 매출 합계는 {total:,}원, 평균은 {average:,}원이다.")
print(f"최고 매출: {best:,}원, 최저 매출: {worst:,}원, 차이: {best - worst:,}원")
print(average >= 300000 and best >= 400000)
print(sorted_sales[:3])
````
