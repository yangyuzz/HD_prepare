# 정답 확인

먼저 스스로 풀고 비교한다.

### 문제 1. 점수 등급 분류 함수 정답

````python
# 문제 1
scores = [88, 72, 95, 63, 80]

def classify_scores(scores):
    if len(scores) == 0:
        return None
    average = sum(scores) / len(scores)
    if average >= 85:
        grade = "우수"
    elif average >= 70:
        grade = "보통"
    else:
        grade = "개선 필요"
    return {"평균": average, "등급": grade}

print(classify_scores(scores))
````

### 문제 2. 성과 요약 함수로 확장하기 정답

````python
# 문제 2
scores = [85, 72, 91, 68, 88]

def classify_performance(scores):
    if len(scores) == 0:
        return None

    average = sum(scores) / len(scores)

    if average >= 85:
        grade = "우수"
    elif average >= 70:
        grade = "보통"
    else:
        grade = "개선 필요"

    return {
        "점수목록": scores,
        "응시인원": len(scores),
        "평균점수": round(average, 2),
        "최고점수": max(scores),
        "최저점수": min(scores),
        "성과등급": grade,
    }

result = classify_performance(scores)
for key, value in result.items():
    print(f"{key}: {value}")
````

### 문제 3. 지역별 매출 합계 함수 정답

````python
# 문제 3
orders = [
    ("서울", "노트북", 2400000), ("부산", "책상", 900000), ("서울", "커피", 300000),
    ("대전", "모니터", 1200000), ("부산", "간식세트", 480000), ("대전", "키보드", 560000),
]

def summarize_by_region(order_list):
    if len(order_list) == 0:
        return None
    totals = {}
    for region, product, amount in order_list:          # 튜플 언패킹
        totals[region] = totals.get(region, 0) + amount   # 없는 지역은 0에서 시작
    return totals

region_totals = summarize_by_region(orders)
for region, total in region_totals.items():
    print(f"{region}: {total:,}원")

best_region = ""
best_total = 0
for region, total in region_totals.items():
    if total > best_total:
        best_region = region
        best_total = total
print(f"매출 1위 지역: {best_region}")
````

### 문제 4. 안전한 숫자 변환 함수로 원본 데이터 정리하기 정답

````python
# 문제 4
raw = [" 2,140", "1,985", "", "2,310 ", "N/A", "2,072", "1,946"]

def to_number(text):
    cleaned = text.strip().replace(",", "")
    try:
        return float(cleaned)
    except ValueError:            # "", "N/A" 처럼 숫자가 아닌 값
        return None

converted = [to_number(text) for text in raw]
print(converted)

valid = [value for value in converted if value is not None]
print(f"변환 실패: {len(raw) - len(valid)}건")

print(f"개수: {len(valid)}개")
print(f"합계: {sum(valid):,.0f}개")
print(f"평균: {sum(valid) / len(valid):,.1f}개")
print(f"최솟값: {min(valid):,.0f}개")
print(f"최댓값: {max(valid):,.0f}개")
````

### 문제 5. 주문 금액 상위 N개 함수 정답

````python
# 문제 5
records = [("모니터", 3, 280000), ("키보드", 12, 42000), ("마우스", 25, 23000),
           ("헤드셋", 4, 89000), ("웹캠", 6, 65000)]

def top_items(records, n=3):
    # r 에 (제품, 수량, 단가) 튜플이 들어오고, 수량 × 단가를 기준으로 정렬한다
    return sorted(records, key=lambda r: r[1] * r[2], reverse=True)[:n]

for rank, item in enumerate(top_items(records), start=1):
    name, quantity, price = item
    print(f"{rank}위 {name}: {quantity * price:,}원")

print(top_items(records, n=1))
print(top_items([]))              # 빈 리스트를 정렬하고 잘라도 빈 리스트
````
