## 2. 함수 정의와 활용 정답

### 문제 2-1 정답

````python
# 문제 2-1
def yield_rate(good, total):
    return round(good / total * 100, 1)

print(f"A라인 수율: {yield_rate(1940, 2000)}%")
print(f"B라인 수율: {yield_rate(1185, 1250)}%")
````

### 문제 2-2 정답

````python
# 문제 2-2
def discount_price(price, rate=0.1):
    return round(price * (1 - rate))

print(discount_price(50000))                  # rate 는 기본값 0.1
print(discount_price(50000, 0.2))             # 위치 순서대로 전달
print(discount_price(rate=0.15, price=80000)) # 이름으로 전달하면 순서를 바꿔도 된다
````

### 문제 2-3 정답

````python
# 문제 2-3
sales = [92000, 145000, 88000, 220000, 167000, 133000, 205000]

def summarize_sales(values):
    if len(values) == 0:      # 빈 리스트면 평균을 계산하지 않고 None
        return None
    return {
        "개수": len(values),
        "합계": sum(values),
        "평균": sum(values) / len(values),
        "최솟값": min(values),
        "최댓값": max(values),
    }

summary = summarize_sales(sales)
for key, value in summary.items():
    if type(value) == float:
        print(f"{key}: {value:,.2f}")
    else:
        print(f"{key}: {value:,}")

print(summarize_sales([]))
````

### 문제 2-4 정답

````python
# 문제 2-4
def sales_range(values):
    low = min(values)
    high = max(values)
    return low, high, high - low      # 쉼표로 여러 값을 반환하면 튜플로 묶인다

result = sales_range([412000, 389000, 455000, 301000])
print(result)
print(type(result))

low, high, gap = sales_range([412000, 389000, 455000, 301000])
print(f"최저 {low:,}원, 최고 {high:,}원, 차이 {gap:,}원")
````

````python
# 문제 2-9
def count_at_least(scores, cutoff):
    count = 0
    for score in scores:
        if score >= cutoff:
            count += 1
    return count

scores = [65, 90, 80, 45, 95]
print(f"80점 이상: {count_at_least(scores, 80)}명")
print(f"90점 이상: {count_at_least(scores, 90)}명")
````

### 문제 2-10 정답

````python
# 문제 2-10
def shipping_fee(amount, remote):
    fee = 0 if amount >= 50000 else 3000
    if remote == "Y":
        fee += 2000
    return fee

amount = int(input("상품 금액: "))
remote = input("도서·산간 여부(Y/N): ")
fee = shipping_fee(amount, remote)
print(f"배송비: {fee:,}원")
print(f"총 결제금액: {amount + fee:,}원")
````

## 3. 예외 처리 정답

### 문제 3-1 정답

````python
# 문제 3-1
budget = 1000000
inputs = ["3200", "1,250", "0"]

for text in inputs:
    try:
        quantity = int(text)
        per_item = budget / quantity
    except ValueError:
        print(f"{text}: 숫자로 변환할 수 없다")
    except ZeroDivisionError:
        print(f"{text}: 0으로 나눌 수 없다")
    else:                           # 오류가 없을 때만 실행
        print(f"{text}: 1개당 예산 {per_item:,.2f}원")
````

### 문제 3-2 정답

````python
# 문제 3-2
try:
    age = int(input("만 나이: "))
except ValueError:
    print("숫자를 입력해야 한다")
else:
    if age < 0:
        print("나이는 0 이상이어야 한다")
    elif age >= 19:
        print("성인")
    else:
        print("미성년자")
````

### 문제 3-3 정답

````python
# 문제 3-3
budget = 10000
try:
    people = int(input("참석 인원: "))
    cost = budget / people
except ValueError:
    print("인원은 숫자로 입력한다")
except ZeroDivisionError:
    print("0명으로 나눌 수 없다")
else:
    print(f"1인당 금액: {cost:.2f}원")
````
