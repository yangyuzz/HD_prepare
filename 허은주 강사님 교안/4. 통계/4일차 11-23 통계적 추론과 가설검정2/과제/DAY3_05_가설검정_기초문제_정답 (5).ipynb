### 1. 일표본 t검정 : 고객센터 연결 대기시간

```python
wait_minutes = np.array([3.8, 4.2, 3.9, 4.4, 3.7, 4.1, 4.0, 3.6, 4.3, 3.9, 4.2, 4.0])

# 귀무가설(H₀): 모집단의 평균 연결 대기시간은 4분이다. (μ = 4)
# 대립가설(H₁): 모집단의 평균 연결 대기시간은 4분과 다르다. (μ ≠ 4)

print("표본평균:", round(wait_minutes.mean(), 3), "분")
print("Shapiro-Wilk p값:", round(stats.shapiro(wait_minutes).pvalue, 4))

result1 = stats.ttest_1samp(wait_minutes, popmean=4)
print("t:", round(result1.statistic, 3), "/ p값:", round(result1.pvalue, 4))

# 정규성 : p = 0.982 ≥ 0.05 → 일표본 t검정을 쓴다
# 통계적 결론 : p = 0.9076 ≥ 0.05 이므로 귀무가설을 기각하지 못한다.
# 사실적 결론 : 모집단의 평균 연결 대기시간이 4분과 다르다고 볼 통계적 근거가 부족하다 (표본평균 4.008분, n = 12).
```

### 2. 대응표본 t검정 : 업무 자동화 전·후 주문 입력 시간

```python
before_work = np.array([15.2, 17.1, 16.4, 20.3, 18.0, 16.6, 19.2, 15.5, 21.1, 17.4])
after_work = np.array([14.1, 15.6, 15.3, 18.2, 16.9, 15.8, 17.0, 14.6, 19.5, 16.2])

# 귀무가설(H₀): 자동화 전·후 주문 입력 시간의 모집단 평균 차이(후 − 전)는 0이다. (μd = 0)
# 대립가설(H₁): 자동화 전·후 주문 입력 시간의 모집단 평균 차이(후 − 전)는 0이 아니다. (μd ≠ 0)

work_diff = after_work - before_work
print("평균 차이(후 − 전):", round(work_diff.mean(), 2), "분")
print("차이값 Shapiro-Wilk p값:", round(stats.shapiro(work_diff).pvalue, 4))

result2 = stats.ttest_rel(after_work, before_work)
print("t:", round(result2.statistic, 3), "/ p값:", result2.pvalue)

# 정규성 : 차이값 p = 0.1499 ≥ 0.05 → 대응표본 t검정을 쓴다
# 통계적 결론 : p = 9.0e-06 (p < 0.001) < 0.05 이므로 귀무가설을 기각한다.
# 사실적 결론 : 자동화 전·후 주문 입력 시간의 평균 차이는 0이 아니라고 볼 통계적 근거가 있다. 표본에서 자동화 후 입력 시간이 평균 1.36분 줄었다 (n = 10명).
```

### 3. Welch t검정 : 두 물류센터의 포장 시간

```python
packing_a = np.array([12, 14, 13, 11, 15, 12, 14, 13, 11, 12])
packing_b = np.array([14, 17, 16, 15, 18, 13, 16, 15, 17, 14, 16])

# 귀무가설(H₀): A와 B 물류센터의 모집단 평균 포장 시간은 같다. (μA = μB)
# 대립가설(H₁): A와 B 물류센터의 모집단 평균 포장 시간은 다르다. (μA ≠ μB)

print("A 평균:", round(packing_a.mean(), 2), "분 / B 평균:", round(packing_b.mean(), 2), "분")
print("A Shapiro-Wilk p값:", round(stats.shapiro(packing_a).pvalue, 4), "/ B Shapiro-Wilk p값:", round(stats.shapiro(packing_b).pvalue, 4))

result3 = stats.ttest_ind(packing_a, packing_b, equal_var=False)
print("t:", round(result3.statistic, 3), "/ p값:", round(result3.pvalue, 4))

# 정규성 : A p = 0.4664, B p = 0.8442 → 둘 다 0.05 이상이므로 Welch t검정을 쓴다
# 통계적 결론 : p = 0.0002 < 0.05 이므로 귀무가설을 기각한다.
# 사실적 결론 : A와 B 물류센터의 모집단 평균 포장 시간은 다르다고 볼 통계적 근거가 있다. 표본에서 A의 평균 포장 시간이 B보다 2.85분 짧았다 (A 10건 · B 11건).
```

### 4. Wilcoxon 부호순위검정 : 재고조사 방식 개선 전·후

```python
before_stock = np.array([58, 62, 55, 70, 63, 81, 59, 68, 73, 65, 66, 60])
after_stock = np.array([55, 60, 51, 67, 61, 78, 55, 65, 71, 61, 53, 67])

# 귀무가설(H₀): 개선 전·후 재고조사 시간 차이(후 − 전)의 중앙값은 0이다. (Md = 0)
# 대립가설(H₁): 개선 전·후 재고조사 시간 차이(후 − 전)의 중앙값은 0이 아니다. (Md ≠ 0)

stock_diff = after_stock - before_stock
print("차이값:", stock_diff)
print("평균 차이:", round(stock_diff.mean(), 2), "분 / 차이의 중앙값:", np.median(stock_diff), "분")
print("차이값 Shapiro-Wilk p값:", round(stats.shapiro(stock_diff).pvalue, 4))

result4 = stats.wilcoxon(after_stock, before_stock)
print("Wilcoxon 통계량:", result4.statistic, "/ p값:", round(result4.pvalue, 4))

# 정규성 : 차이값 p = 0.0025 < 0.05 → 양쪽 끝(−13분, +7분)에 멀리 떨어진 매장이 있어 정규성이 어긋난다
# 대칭 : 차이값이 −3분을 중심으로 양쪽에 비슷하게 놓여 있다 (−13분과 +7분이 각각 한쪽 끝, 나머지는 −4 ~ −2분). 평균과 중앙값도 −3분으로 같다 → Wilcoxon 부호순위검정을 쓴다
# 통계적 결론 : p = 0.0269 < 0.05 이므로 귀무가설을 기각한다.
# 사실적 결론 : 개선 전·후 재고조사 시간 차이의 중앙값은 0이 아니라고 볼 통계적 근거가 있다. 표본에서 12곳 중 11곳의 시간이 줄었고 차이의 중앙값은 −3분이다 (n = 12곳).
```

### 5. Mann-Whitney U 검정 : 두 배송업체의 고객 만족도

```python
delivery_a = np.array([4, 5, 4, 3, 5, 4, 4, 5, 4, 5, 4, 3])
delivery_b = np.array([3, 3, 4, 2, 3, 2, 4, 3, 3, 2, 3, 4])

# 귀무가설(H₀): A와 B 배송업체 고객 만족도의 분포는 같다. (F₁ = F₂)
# 대립가설(H₁): A와 B 배송업체 고객 만족도의 분포는 다르다. (F₁ ≠ F₂)

print("A 중앙값:", np.median(delivery_a), "점 / B 중앙값:", np.median(delivery_b), "점")
print("A Shapiro-Wilk p값:", round(stats.shapiro(delivery_a).pvalue, 4), "/ B Shapiro-Wilk p값:", round(stats.shapiro(delivery_b).pvalue, 4))

result5 = stats.mannwhitneyu(delivery_a, delivery_b)
print("U:", result5.statistic, "/ p값:", round(result5.pvalue, 4))

# 정규성 : A p = 0.0153, B p = 0.0199 → 둘 다 0.05 미만이므로 Mann-Whitney U 검정을 쓴다
# 통계적 결론 : p = 0.0022 < 0.05 이므로 귀무가설을 기각한다.
# 사실적 결론 : A와 B 배송업체 고객 만족도의 분포는 다르다고 볼 통계적 근거가 있다. 표본에서 A의 만족도가 더 높게 나오는 경향이 있다 (중앙값 A 4점 · B 3점, 각 12명).
```

## 개념 확인 정답

### 개념 1. p값의 뜻으로 알맞은 것은?

**답 :** ②

p값은 귀무가설이 참이라고 가정하고 계산한 확률이다. 귀무가설이나 대립가설이 참일 확률이 아니다.

### 개념 2. 유의수준 α = 0.05의 뜻으로 알맞은 것은?

**답 :** ①

유의수준은 귀무가설이 참인데 기각하는 잘못(제1종 오류)을 어디까지 허용할지 데이터를 보기 전에 정한 기준이다.

### 개념 3. p = 0.30이 나왔다. 알맞은 결론은?

**답 :** ③

p ≥ 0.05이면 귀무가설을 기각하지 못한다. 귀무가설이 참임을 입증한 것은 아니다.

### 개념 4. 대응표본 t검정에서 정규성을 확인하는 대상은?

**답 :** ④

대응표본 t검정은 차이값의 평균이 0인지 검정하므로 차이값의 분포를 확인한다.

### 개념 5. 서로 다른 두 집단의 평균을 비교할 때 기본으로 쓰는 함수는?

**답 :** ③

독립된 두 집단은 Welch t검정(`equal_var=False`)을 기본으로 쓴다. ①은 일표본, ②는 대응표본, ④는 대응표본의 비모수 검정이다.

### 개념 6. 독립된 두 집단 중 어느 집단의 값이 더 크게 나오는 경향인지 비교하려 한다. 표본이 작고 한 집단의 분포가 한쪽으로 크게 치우쳤다. 알맞은 검정은?

**답 :** ②

비교 목적이 어느 집단의 값이 더 크게 나오는 경향인지이고, 표본이 작으며 한 집단의 정규성이 크게 어긋나므로 비모수 검정인 Mann-Whitney U 검정을 쓴다. 검정은 정규성 p값 하나가 아니라 비교할 대상, 표본 크기, 분포 모양을 함께 보고 고른다.

### 개념 7. 95% 신뢰구간의 뜻으로 알맞은 것은?

**답 :** ②

신뢰구간의 95%는 구간을 만드는 방법의 신뢰도다. 개별 데이터가 들어가는 범위가 아니다.

### 개념 8. 표본 크기 n이 커질 때 표준오차(s ÷ √n)는 어떻게 되는가?

**답 :** ②

표준오차는 √n으로 나누므로 표본이 커질수록 작아진다. 표본평균이 모평균 가까이에 더 좁게 모인다.

### 개념 9. 검정통계량의 뜻으로 알맞은 것은?

**답 :** ②

검정통계량은 표본 데이터로 계산하여 관측 결과가 귀무가설에서 얼마나 벗어났는지를 하나의 숫자로 나타낸 값이다. p값은 이 값으로 계산한다. 검정마다 계산 방법과 이름이 다르다 : t검정은 t, Wilcoxon 부호순위검정은 순위합 통계량, Mann-Whitney U 검정은 U. 예를 들어 일표본 t검정의 t = (표본평균 − 기준값) ÷ 표준오차는 표본평균이 기준값에서 표준오차 몇 개만큼 떨어져 있는지를 뜻한다. ①은 표본평균, ③은 임계값, ④는 표준편차다.

### 개념 10. 대응표본으로 분석해야 하는 경우는?

**답 :** ②

같은 대상을 두 번 측정해 관측값이 짝을 이루면 대응표본이다. ①과 ③은 서로 다른 대상의 두 집단이므로 독립표본, ④는 일표본이다.
