## 2. AI4I 토크의 대표값

`Torque [Nm]`는 설비의 토크 측정값이며 단위는 Nm(뉴턴미터)이다. 대표값은 유효값 수(`count`)와 단위를 함께 적어야 해석할 수 있다.

### **`agg()`의 의미**

`agg()`는 **데이터에 합계, 평균, 최댓값 등 하나 또는 여러 집계 함수를 적용할 때** 사용한다.

### **기본 문법**

```python
# 1. 하나 또는 여러 집계 함수 적용
data.agg(["sum", "mean", "max"])

# 2. 그룹별로 한 열을 선택한 경우
df.groupby("지역")["매출"].agg(
    매출합계="sum",
    매출평균="mean"
)

# 3. 그룹별로 여러 열을 계산하는 경우
df.groupby("지역").agg(
    매출합계=("매출", "sum"),
    수량평균=("수량", "mean")
)
```

### **사용 방법 비교**

| 구분 | 형식 | 의미 |
|---|---|---|
| 일반 집계 | `agg(["sum", "mean"])` | 여러 집계 함수 적용 |
| 한 열의 그룹별 집계 | `agg(새열이름="함수명")` | 이미 선택한 열을 집계 |
| 여러 열의 그룹별 집계 | `agg(새열이름=("열이름", "함수명"))` | 계산할 열과 함수 지정 |



공식 문서: [DataFrame.agg()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.agg.html) · [Pandas 그룹별 집계](https://pandas.pydata.org/docs/user_guide/groupby.html#named-aggregation)

## 3. 범위와 IQR

| 지표 | 계산 | 의미 |
|---|---|---|
| 범위 | 최댓값 − 최솟값 | 전체 폭 |
| Q1 | 25% 분위수 | 아래쪽 25% 위치의 기준 |
| Q3 | 75% 분위수 | 아래쪽 75% 위치의 기준 |
| IQR | Q3 − Q1 | 가운데 50%의 퍼짐 |

IQR은 범위보다 극단값에 상대적으로 덜 민감하다. 작은 자료나 동점이 많은 자료에서는 분위수 사이에 정확히 절반의 행이 들어가지 않을 수 있다.

`quantile()`은 지정한 분위수 위치의 값을 계산한다. 0.25는 Q1, 0.50은 중앙값, 0.75는 Q3이다. 가운데 부분의 폭을 보려면 `IQR = Q3 - Q1`을 계산한다.

전체 폭과 가운데 부분의 폭을 따로 구하면 양 끝값에 따른 퍼짐과 자료 중심부의 퍼짐을 구분할 수 있다.

공식 문서 : [quantile](https://pandas.pydata.org/docs/reference/api/pandas.Series.quantile.html)

### **`quantile()`의 의미**

`quantile()`은 **데이터를 작은 값부터 큰 값 순서로 정렬했을 때, 특정 비율에 해당하는 값(분위수)을 구하는 함수**이다.

데이터의 25%, 50%, 75% 지점을 확인할 때 주로 사용한다.

| 코드 | 의미 |
|---|---|
| `quantile(0.25)` | 25% 지점의 값 (1사분위수, Q1) |
| `quantile(0.5)` | 50% 지점의 값 (중앙값, Q2) |
| `quantile(0.75)` | 75% 지점의 값 (3사분위수, Q3) |

### **기본 사용법**

```python
import pandas as pd

values = pd.Series([10, 20, 30, 40, 50])

print(values.quantile(0.25))  # 20.0
print(values.quantile(0.5))   # 30.0
print(values.quantile(0.75))  # 40.0
```

여러 분위수를 한 번에 계산할 수도 있다.

```python
print(values.quantile([0.25, 0.5, 0.75]))
```

Q1과 Q3는 데이터의 가운데 50%가 분포하는 범위를 확인할 때 사용한다.

두 값의 차이를 **사분위범위(IQR)**라고 한다.

```python
Q1 = torque.quantile(0.25)
Q3 = torque.quantile(0.75)

IQR = Q3 - Q1
```

**핵심 정리**

- `quantile(0.25)` : 1사분위수(Q1)
- `quantile(0.5)` : 중앙값(Q2)
- `quantile(0.75)` : 3사분위수(Q3)
- `Q3 - Q1` : 사분위범위(IQR)

공식 문서: [Pandas Series.quantile()](https://pandas.pydata.org/docs/reference/api/pandas.Series.quantile.html)

## 4. 편차·분산·표준편차

편차는 값−평균이다. 편차의 제곱을 모아 분산을 구하고, 분산의 제곱근으로 표준편차를 구한다. 분산의 단위는 (Nm)², 표준편차는 Nm이다.

평균만으로는 흩어진 정도를 알 수 없으므로 각 값의 편차를 모아 퍼짐을 계산한다. 양수와 음수 편차가 서로 상쇄되지 않도록 제곱하고, 원래 단위로 읽으려면 제곱근을 사용한다.

### **`var()`와 `std()`의 의미**

분산과 표준편차는 **데이터가 평균을 중심으로 얼마나 퍼져 있는지** 나타내는 통계량이다.

- **`var()` (분산)**: 각 데이터가 평균에서 떨어진 정도를 제곱하여 계산한다.
- **`std()` (표준편차)**: 분산에 제곱근을 적용한 값이다.

두 값 모두 클수록 데이터가 평균을 중심으로 넓게 퍼져 있다는 의미이다.

### **기본 사용법**

```python
series.var(ddof=1, skipna=True)
series.std(ddof=1, skipna=True)
```

| 항목 | 기본값 | 의미 |
|---|---|---|
| `ddof` | 1 | 분산과 표준편차 계산 시 나누는 수를 조정 |
| `skipna` | True | 결측값(NaN)을 제외하고 계산 |

### **`ddof=0`과 `ddof=1`의 차이**

- **`ddof=0`**: 유효 데이터 개수 N으로 나눈다. (모집단 분산·표준편차)
- **`ddof=1`**: 유효 데이터 개수 N-1로 나눈다. (표본 분산·표준편차)

Pandas는 기본적으로 `ddof=1`을 사용한다.

### **실행 예제**

```python
import pandas as pd

example = pd.Series([10, 20, 30])

# 1. 평균 계산
print("평균:", example.mean())  # 20.0

# 2. 모집단 분산과 표준편차 (ddof=0)
print("모집단 분산:", round(example.var(ddof=0), 2))   # 66.67
print("모집단 표준편차:", round(example.std(ddof=0), 2))  # 8.16

# 3. 표본 분산과 표준편차 (ddof=1)
print("표본 분산:", round(example.var(ddof=1), 2))   # 100.0
print("표본 표준편차:", round(example.std(ddof=1), 2))  # 10.0
```

**계산 결과 비교**

| 구분 | `ddof=0` | `ddof=1` |
|---|---:|---:|
| 분산 `var()` | 66.67 | 100.00 |
| 표준편차 `std()` | 8.16 | 10.00 |
| 나누는 수 | N = 3 | N−1 = 2 |

### **분산과 표준편차의 차이**

분산은 평균과의 차이를 제곱하여 계산하므로 **원래 데이터 단위의 제곱**을 사용한다.

표준편차는 분산에 제곱근을 적용하므로 **원래 데이터와 같은 단위**를 사용한다.

예를 들어 온도 데이터의 단위가 °C라면 분산의 단위는 °C²이고, 표준편차의 단위는 °C이다.

**핵심 정리**

- `var()` : 분산 계산
- `std()` : 표준편차 계산
- `ddof=0` : 전체 데이터를 모집단으로 보고 계산
- `ddof=1` : 표본 분산과 표본 표준편차 계산 (기본값)
- `skipna=True` : 결측값 제외 (기본값)

공식 문서: [Pandas Series.var()](https://pandas.pydata.org/docs/reference/api/pandas.Series.var.html) · [Pandas Series.std()](https://pandas.pydata.org/docs/reference/api/pandas.Series.std.html)

````python
# [예제] ddof=0과 ddof=1 비교
# 문제: 데이터 [2, 4, 6]의 평균과 편차를 구하고,
#       ddof=0과 ddof=1의 분산 및 표준편차를 비교한다.

example = pd.Series([2, 4, 6])


# 1) 평균과 편차 계산
# 평균: (2 + 4 + 6) / 3 = 4
# 편차: 각 데이터 - 평균 → [-2, 0, 2]

print("평균:", example.mean())
print("편차:", (example - example.mean()).tolist())


# 2) ddof=0 : 모집단 분산과 표준편차
#    var : 분산 / std : 표준편차
#    데이터 개수 N=3으로 나눈다.
#    편차 제곱 합계 = 4 + 0 + 4 = 8
#    분산 = 8 / 3 = 2.6667

print("분산(ddof=0):", round(example.var(ddof=0), 4))
print("표준편차(ddof=0):", round(example.std(ddof=0), 4))


# 3) ddof=1 : 표본 분산과 표준편차
#    데이터 개수 N-1=2로 나눈다.
#    분산 = 8 / 2 = 4

print("분산(ddof=1):", round(example.var(ddof=1), 4))
print("표준편차(ddof=1):", round(example.std(ddof=1), 4))
````

## **5. 변동계수(CV, Coefficient of Variation)**

변동계수(CV)는 **평균에 비해 데이터가 얼마나 퍼져 있는지 백분율(%)로 나타낸 값**이다.

표준편차는 데이터의 퍼짐 정도를 나타내고, CV는 평균을 기준으로 상대적인 퍼짐 정도를 나타낸다. 평균이나 규모가 다른 데이터의 변동 정도를 비교할 때 사용한다.

### **계산 공식**

**CV(%) = (표준편차 ÷ 평균) × 100**

예를 들어 평균이 100이고 표준편차가 10이라면,

```text
CV = (10 ÷ 100) × 100 = 10%
```

즉, **표준편차가 평균의 10%에 해당한다**는 의미이다.

### **CV 결과 해석**

| 구분 | 평균 | 표준편차 | CV |
|---|---:|---:|---:|
| A 설비 | 100 | 10 | 10% |
| B 설비 | 200 | 40 | 20% |

- A 설비: CV 10% → 평균에 비해 상대적인 변동이 작다.
- B 설비: CV 20% → 평균에 비해 상대적인 변동이 크다.

따라서 이 예제에서는 **A 설비가 B 설비보다 상대적인 변동이 작다.**

### **Pandas 계산 예제**

```python
import pandas as pd

values = pd.Series([90, 100, 110])

# 1) 평균과 표준편차 계산
mean_value = values.mean()
std_value = values.std(ddof=1)

# 2) 변동계수 계산
cv = std_value / mean_value * 100

print("평균:", mean_value)
print("표준편차:", round(std_value, 2))
print("변동계수:", round(cv, 2), "%")
```

**실행 결과**

```text
평균: 100.0
표준편차: 10.0
변동계수: 10.0 %
```

※ 실습 자료는 표본이므로 CV는 `ddof=1`로 계산한다.

### **CV 사용 시 주의할 점**

**① 0이 실제로 측정량이 없음을 의미하는 데이터에 사용한다.**

| 데이터 | 0의 의미 | CV 사용 |
|---|---|---|
| 생산량 0개 | 생산한 제품이 없음 | 적절 |
| 무게 0kg | 측정량이 없음 | 적절 |
| 길이 0cm | 길이가 없음 | 적절 |
| 섭씨온도 0°C | 온도가 없다는 뜻이 아님 | 일반적으로 부적절 |

생산량 200개는 100개의 두 배라고 할 수 있지만, 섭씨 20°C를 10°C의 두 배라고 해석할 수는 없다.

따라서 **0이 측정량이 없다는 것을 의미하는 비율척도 데이터**에서 CV를 사용하는 것이 적절하다.

**② 평균이 0이거나 0에 가까우면 주의한다.**

CV는 평균으로 나누어 계산하므로 평균이 0이면 계산할 수 없고, 0에 가까우면 CV가 지나치게 커져 해석하기 어려울 수 있다.

**③ CV가 작다고 반드시 좋은 것은 아니다.**

CV가 작다는 것은 상대적인 변동이 작다는 의미이지, 설비나 공정의 품질이 반드시 좋다는 뜻은 아니다. 분석 목적과 품질 기준을 함께 확인해야 한다.

### **정리**

- **표준편차(std)**: 데이터가 평균에서 얼마나 퍼져 있는가?
- **변동계수(CV)**: 평균에 비해 얼마나 퍼져 있는가?
- **CV가 작을수록** 평균 대비 상대적인 변동이 작다.
- **CV가 클수록** 평균 대비 상대적인 변동이 크다.
- 평균이 0에 가깝거나 의미 있는 0이 없는 데이터에서는 사용에 주의한다.

## 6. describe() 출력 읽기

수치형 describe()의 count는 유효 관측수, std는 ddof=1 표준편차, 50%는 중앙값이다. 결측값 수·분산·범위·IQR·분포 모양은 별도로 확인한다.

| 출력 항목 | 의미 |
|---|---|
| count | 유효 관측값 수 |
| mean | 평균 |
| std | 표준편차 |
| min | 최솟값 |
| 25% | Q1 |
| 50% | 중앙값 |
| 75% | Q3 |

## 7. 정리

| 확인할 것 | 함수·계산 | 답하는 질문 |
|---|---|---|
| 평균 | `mean()` | 값들의 중심은 얼마인가 (극단값에 민감) |
| 중앙값 | `median()` | 정렬했을 때 가운데 값은 얼마인가 (극단값에 덜 민감) |
| 여러 통계량 한 번에 | `agg(["count", "mean", ...])` | 필요한 요약값만 골라 한 번에 |
| 범위 | `max() - min()` | 전체 값이 퍼진 폭은 얼마인가 |
| 분위수 | `quantile(0.25)`, `quantile(0.75)` | 아래에서 25%, 75% 위치의 값은 얼마인가 |
| IQR | `Q3 - Q1` | 가운데 50% 가 퍼진 폭은 얼마인가 |
| 분산 | `var(ddof=1)` | 평균에서 떨어진 정도 (단위의 제곱) |
| 표준편차 | `std(ddof=1)` | 평균에서 떨어진 정도 (원래 단위) |
| 변동계수 | `std(ddof=1) / mean() * 100` | 평균에 비해 얼마나 퍼져 있나 (%) |
| 한 번에 요약 | `describe()` | 개수·평균·표준편차·최솟값·분위수·최댓값 |

순서
1) 중심을 본다 : 평균과 중앙값을 같이 보고, 둘이 크게 다르면 극단값이나 치우침을 의심한다
2) 퍼짐을 본다 : 범위와 IQR, 표준편차를 같이 본다
3) 평균 규모가 다른 자료의 퍼짐을 비교할 때는 CV 를 쓴다. 0 이 '없음' 을 뜻하는 자료에만 쓴다
4) 데이터와 요약 결과는 다른 변수에 담는다

공식 문서 : [mean](https://pandas.pydata.org/docs/reference/api/pandas.Series.mean.html) · [median](https://pandas.pydata.org/docs/reference/api/pandas.Series.median.html) · [agg](https://pandas.pydata.org/docs/reference/api/pandas.Series.agg.html) · [quantile](https://pandas.pydata.org/docs/reference/api/pandas.Series.quantile.html) · [var](https://pandas.pydata.org/docs/reference/api/pandas.Series.var.html) · [std](https://pandas.pydata.org/docs/reference/api/pandas.Series.std.html) · [describe](https://pandas.pydata.org/docs/reference/api/pandas.Series.describe.html)
